```python
import pickle
import pandas as pd
```

```python
def load_model(model_path='models/best_linear_model.pkl'):
    """저장된 모델과 scaler 로드"""
    with open(model_path, 'rb') as f:
        saved = pickle.load(f)
        model = saved['model']
        scaler = saved['scaler']
    return model, scaler


def load_and_preprocess_test(test_path='models/mission15_test.csv'):
    """테스트 데이터 로드 및 전처리"""
    test_df = pd.read_csv(test_path)
    test_df['Extracurricular Activities'] = test_df['Extracurricular Activities'].apply(
        lambda x: 1 if x == 'Yes' else 0
    )
    return test_df


def predict(model, scaler, X_test):
    """데이터 정규화 및 예측"""
    X_test_scaled = scaler.transform(X_test)
    predictions = model.predict(X_test_scaled)
    return predictions


def save_predictions(predictions, output_path='result.csv'):
    """예측 결과 저장"""
    result_df = pd.DataFrame({'Performance Index': predictions})
    result_df.to_csv(output_path, index=False)
    print(f"예측 결과 저장 완료: {output_path}")


def main():
    # 모델 로드
    model, scaler = load_model()
    print("모델 로드 완료")
    
    # 테스트 데이터 로드 및 전처리
    test_df = load_and_preprocess_test()
    print(f"테스트 데이터 로드 완료: {len(test_df)}개 샘플")
    
    # 예측
    predictions = predict(model, scaler, test_df)
    print("예측 결과 (처음 10개):")
    print(predictions[:10])
    
    # 결과 저장
    save_predictions(predictions)
```

```python
if __name__ == "__main__":
    main()
```
