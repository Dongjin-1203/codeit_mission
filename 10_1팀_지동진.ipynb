### 데이터 전처리(문장길이 전처리 코드 2가지 모두 첨부)

````python
def clean_text_with_nltk(text):
    # 소문자 변환
    text = text.lower()

    # 특수문자 제거 (선택사항 - tokenize가 알아서 처리하기도 함)
    text = re.sub(r'[^a-zA-Z\s]', '', text)

    # NLTK word_tokenize 사용
    tokens = word_tokenize(text)

    # 불용어 제거
    stop_words = set(stopwords.words('english'))
    tokens = [word for word in tokens if word not in stop_words]

    return ' '.join(tokens)
````

````python
# 문장 길이 기준으로 데이터 필터링
def filter_by_length(texts, labels, min_len=10, max_len=500):

    from nltk.tokenize import word_tokenize

    filtered_texts = []
    filtered_labels = []

    removed_too_short = 0
    removed_too_long = 0

    for text, label in zip(texts, labels):
        # 단어 수 계산
        word_count = len(word_tokenize(text))

        # 길이 체크
        if word_count < min_len:
            removed_too_short += 1
            continue
        elif word_count > max_len:
            removed_too_long += 1
            continue

        filtered_texts.append(text)
        filtered_labels.append(label)

    print(f"\n{'='*70}")
    print("문장 길이 필터링 결과")
    print(f"{'='*70}")
    print(f"원본 데이터:           {len(texts):,}개")
    print(f"너무 짧음 제거 (<{min_len}):  {removed_too_short:,}개")
    print(f"너무 김 제거 (>{max_len}):    {removed_too_long:,}개")
    print(f"최종 데이터:           {len(filtered_texts):,}개")
    print(f"유지 비율:             {len(filtered_texts)/len(texts)*100:.1f}%")
    print(f"{'='*70}\n")

    return filtered_texts, np.array(filtered_labels)
````

````python
# 데이터에 적용
X_train_processed = [clean_text_with_nltk(text) for text in X_train]
X_test_processed = [clean_text_with_nltk(text) for text in X_test]
````

````python
# ✅ 문장 길이 필터링 추가!
X_train_filtered, y_train_filtered = filter_by_length(
    X_train_processed, y_train,
    min_len=15,    # 최소 15단어
    max_len=400    # 최대 400단어
)

X_test_filtered, y_test_filtered = filter_by_length(
    X_test_processed, y_test,
    min_len=15,
    max_len=400
)
````

````python
class MultiEmbedding:
    """Word2Vec, FastText를 한 번에 학습/변환하는 통합 클래스"""

    def __init__(self,
                 vector_size=100,
                 window=5,
                 min_count=5,
                 sg=1,
                 epochs=10,
                 workers=4,
                 max_features=10000,
                 min_df=5,
                 seed=156,
                 embeddings=['word2vec', 'fasttext']):

        self.vector_size = vector_size
        self.window = window
        self.min_count = min_count
        self.sg = sg
        self.epochs = epochs
        self.workers = workers
        self.max_features = max_features
        self.min_df = min_df
        self.seed = seed
        self.embeddings = embeddings

        # 모델 저장
        self.models = {}
        self.is_trained = False

    def fit(self, X_train):
        """모든 임베딩 모델 학습"""
        print("\n" + "="*70)
        print("통합 임베딩 학습 시작")
        print("="*70)

        if 'word2vec' in self.embeddings:
            self._train_word2vec(X_train)

        if 'fasttext' in self.embeddings:
            self._train_fasttext(X_train)

        self.is_trained = True
        print("\n" + "="*70)
        print("모든 임베딩 학습 완료!")
        print("="*70 + "\n")

        return self

    def _train_word2vec(self, X_train):
        """Word2Vec 학습"""
        print("\n[1/3] Word2Vec 학습 중...")
        print("-" * 70)

        train_sentences = [text.split() for text in X_train]

        self.models['word2vec'] = Word2Vec(
            sentences=train_sentences,
            vector_size=self.vector_size,
            window=self.window,
            min_count=self.min_count,
            workers=self.workers,
            sg=self.sg,
            epochs=self.epochs,
            seed=self.seed
        )

        print(f"✓ Word2Vec Vocabulary 크기: {len(self.models['word2vec'].wv)}")

    def _train_fasttext(self, X_train):
        """FastText 학습"""
        print("\n[2/3] FastText 학습 중...")
        print("-" * 70)

        train_sentences = [text.split() for text in X_train]

        self.models['fasttext'] = FastText(
            sentences=train_sentences,
            vector_size=self.vector_size,
            window=self.window,
            min_count=self.min_count,
            workers=self.workers,
            sg=self.sg,
            epochs=self.epochs,
            seed=self.seed
        )

        print(f"✓ FastText Vocabulary 크기: {len(self.models['fasttext'].wv)}")

    def transform(self, X):
        """모든 임베딩으로 문서 변환"""
        if not self.is_trained:
            raise ValueError("모델이 학습되지 않았습니다. fit()을 먼저 호출하세요.")

        print("\n" + "="*70)
        print("데이터 벡터화 진행")
        print("="*70)

        results = {}

        if 'word2vec' in self.embeddings:
            results['word2vec'] = self._transform_word2vec(X)

        if 'fasttext' in self.embeddings:
            results['fasttext'] = self._transform_fasttext(X)

        print("\n✓ 모든 벡터화 완료!\n")

        return results

    def _transform_word2vec(self, X):
        """Word2Vec 변환"""
        print("\n[Word2Vec 벡터화]")
        model = self.models['word2vec']

        def document_vector(doc):
            if isinstance(doc, str):
                doc = doc.split()
            vectors = [model.wv[word] for word in doc if word in model.wv]
            return np.mean(vectors, axis=0) if vectors else np.zeros(self.vector_size)

        vectors = np.array([document_vector(doc) for doc in X])
        print(f"  Shape: {vectors.shape}")
        return vectors

    def _transform_fasttext(self, X):
        """FastText 변환"""
        print("\n[FastText 벡터화]")
        model = self.models['fasttext']

        def document_vector(doc):
            if isinstance(doc, str):
                doc = doc.split()
            vectors = [model.wv[word] for word in doc]
            return np.mean(vectors, axis=0) if vectors else np.zeros(self.vector_size)

        vectors = np.array([document_vector(doc) for doc in X])
        print(f"  Shape: {vectors.shape}")
        return vectors

    def fit_transform(self, X_train):
        """학습 + 변환"""
        self.fit(X_train)
        return self.transform(X_train)

    def save_models(self, prefix='./'):
        """모든 모델 저장"""
        print("\n" + "="*70)
        print("모델 저장 중")
        print("="*70)

        if 'word2vec' in self.models:
            filepath = f'{prefix}word2vec_model.model'
            self.models['word2vec'].save(filepath)
            print(f"✓ Word2Vec 저장: {filepath}")

        if 'fasttext' in self.models:
            filepath = f'{prefix}fasttext_model.model'
            self.models['fasttext'].save(filepath)
            print(f"✓ FastText 저장: {filepath}")

        print("✓ 모든 모델 저장 완료!\n")

    def get_vocab_info(self):
        """각 임베딩의 vocabulary 정보 출력"""
        print("\n" + "="*70)
        print("Vocabulary 정보")
        print("="*70)

        if 'word2vec' in self.models:
            print(f"Word2Vec: {len(self.models['word2vec'].wv)} 단어")

        if 'fasttext' in self.models:
            print(f"FastText:  {len(self.models['fasttext'].wv)} 단어")

        print()
````

### 데이터셋 클래스, 데이터 로더 클래스

````python
class TextDataset(Dataset):
    def __init__(self, texts, labels, word2idx, max_len,
                 pad_idx=0, unk_idx=1, lowercase=True, add_offset=True):
        self.texts = texts
        self.labels = labels
        self.lowercase = lowercase
        self.max_len = max_len
        self.pad_idx = pad_idx
        self.unk_idx = unk_idx
        self.add_offset = add_offset

        # ✅ word2idx를 미리 +2 오프셋으로 변환
        if add_offset:
            self.word2idx = {word: idx + 2 for word, idx in word2idx.items()}
        else:
            self.word2idx = word2idx

    def __len__(self):
        return len(self.texts)

    def __getitem__(self, idx):
        text = self.texts[idx]

        if self.lowercase:
            text = text.lower()

        tokens = word_tokenize(text)
        encoded = [self.word2idx.get(word, self.unk_idx) for word in tokens]

        if len(encoded) < self.max_len:
            encoded += [self.pad_idx] * (self.max_len - len(encoded))
        else:
            encoded = encoded[:self.max_len]

        encoded_tensor = torch.tensor(encoded, dtype=torch.long)
        label_tensor = torch.tensor(self.labels[idx], dtype=torch.long)

        return encoded_tensor, label_tensor
````

````python
class EmbeddingDataLoaderManager:
    """3가지 임베딩의 DataLoader를 관리하는 클래스"""

    def __init__(self, embedding, X_train, y_train, X_test, y_test,
                 max_len=256, batch_size=64, pad_idx=0, unk_idx=1):

        self.embedding = embedding
        self.X_train = X_train
        self.y_train = y_train
        self.X_test = X_test
        self.y_test = y_test
        self.max_len = max_len
        self.batch_size = batch_size
        self.pad_idx = pad_idx
        self.unk_idx = unk_idx

        self.dataloaders = {}
        self._create_all_dataloaders()

    def _create_all_dataloaders(self):
        """모든 임베딩에 대한 DataLoader 생성"""
        print("\n" + "="*70)
        print("DataLoader 생성 중...")
        print("="*70)

        # Word2Vec
        if 'word2vec' in self.embedding.models:
            self.dataloaders['word2vec'] = self._create_dataloader('word2vec')

        # FastText
        if 'fasttext' in self.embedding.models:
            self.dataloaders['fasttext'] = self._create_dataloader('fasttext')

        print("\n✓ 모든 DataLoader 생성 완료!\n")

    def _create_dataloader(self, emb_name):
        """특정 임베딩에 대한 DataLoader 생성"""
        # word2idx 추출
        if emb_name in ['word2vec', 'fasttext']:
            word2idx = self.embedding.models[emb_name].wv.key_to_index
        else:
            print("없는 임베딩입니다.")

        # Dataset 생성 (✅ add_offset=True 추가)
        train_dataset = TextDataset(
            self.X_train, self.y_train, word2idx,
            self.max_len, self.pad_idx, self.unk_idx,
            add_offset=True  # ✅ 추가!
        )
        test_dataset = TextDataset(
            self.X_test, self.y_test, word2idx,
            self.max_len, self.pad_idx, self.unk_idx,
            add_offset=True  # ✅ 추가!
        )

        # DataLoader 생성
        train_loader = DataLoader(train_dataset, batch_size=self.batch_size, shuffle=True)
        test_loader = DataLoader(test_dataset, batch_size=self.batch_size, shuffle=False)

        print(f"  ✓ {emb_name.upper():10s} (vocab: {len(word2idx):5d} 단어)")

        return train_loader, test_loader

    def get_dataloader(self, emb_name):
        """특정 임베딩의 DataLoader 가져오기"""
        if emb_name not in self.dataloaders:
            raise ValueError(f"{emb_name}에 대한 DataLoader가 없습니다.")
        return self.dataloaders[emb_name]

    def get_vocab_size(self, emb_name):
        """특정 임베딩의 vocabulary 크기"""
        if emb_name in ['word2vec', 'fasttext']:
            return len(self.embedding.models[emb_name].wv.key_to_index)
        else:
            print("없는 임베딩입니다.")

    def show_info(self):
        """DataLoader 정보 출력"""
        print("\n" + "="*70)
        print("DataLoader 정보")
        print("="*70)

        for emb_name in self.dataloaders.keys():
            train_loader, test_loader = self.dataloaders[emb_name]
            vocab_size = self.get_vocab_size(emb_name)

            print(f"\n[{emb_name.upper()}]")
            print(f"  Vocabulary 크기: {vocab_size}")
            print(f"  Train batches: {len(train_loader)}")
            print(f"  Test batches:  {len(test_loader)}")
            print(f"  Batch size:    {self.batch_size}")
            print(f"  Max length:    {self.max_len}")
````

### 모델 클래스(양방향 LSTM)

````python
class BiLSTMModel(nn.Module):
    def __init__(self, embedding_matrix, hidden_dim, output_dim,
                 n_layers=2, dropout=0.3, freeze_embeddings=False, pooling='last'):
        super(BiLSTMModel, self).__init__()

        self.hidden_dim = hidden_dim
        self.n_layers = n_layers
        self.pooling = pooling

        # 임베딩 레이어
        self.embedding = nn.Embedding.from_pretrained(
            torch.tensor(embedding_matrix, dtype=torch.float),
            freeze=freeze_embeddings
        )

        embedding_dim = embedding_matrix.shape[1]

        # 양방향 LSTM
        self.lstm = nn.LSTM(
            embedding_dim,
            hidden_dim,
            num_layers=n_layers,
            dropout=dropout if n_layers > 1 else 0,
            batch_first=True,
            bidirectional=True  # ✅ 양방향!
        )

        # 양방향이므로 출력 차원 2배
        lstm_output_dim = hidden_dim * 2

        # 드롭아웃
        self.dropout = nn.Dropout(dropout)

        # Fully Connected Layers
        self.fc1 = nn.Linear(lstm_output_dim, 128)  # ✅ 2배 차원
        self.fc2 = nn.Linear(128, 64)
        self.fc3 = nn.Linear(64, output_dim)

    def forward(self, x):
        # 임베딩
        embedded = self.embedding(x)

        # 양방향 LSTM
        lstm_out, (hidden, cell) = self.lstm(embedded)
        # lstm_out: (batch, seq_len, hidden_dim * 2)
        # hidden: (n_layers * 2, batch, hidden_dim)

        # Pooling
        if self.pooling == 'last':
            # Forward와 Backward의 마지막 hidden state concat
            # hidden[-2]: forward의 마지막 레이어
            # hidden[-1]: backward의 마지막 레이어
            output = torch.cat((hidden[-2], hidden[-1]), dim=1)

        elif self.pooling == 'mean':
            # LSTM 출력의 평균
            output = torch.mean(lstm_out, dim=1)

        elif self.pooling == 'max':
            # LSTM 출력의 최댓값
            output = torch.max(lstm_out, dim=1)[0]
        else:
            raise ValueError(f"Unknown pooling: {self.pooling}")

        # FC Layers
        output = self.dropout(output)
        output = F.relu(self.fc1(output))
        output = self.dropout(output)
        output = F.relu(self.fc2(output))
        output = self.dropout(output)
        output = self.fc3(output)

        return output
````

### 모델 학습 및 평가

````python
class EmbeddingModelEvaluator:
    """3가지 임베딩 × 3가지 모델 = 9개 조합 평가 시스템"""

    def __init__(self, embedding_obj, dl_manager, class_names, device='cuda'):

        self.embedding_obj = embedding_obj
        self.dl_manager = dl_manager
        self.class_names = class_names
        self.device = torch.device(device if torch.cuda.is_available() else 'cpu')

        self.embedding_names = ['word2vec', 'fasttext']
        self.model_classes = {'BiLSTM': BiLSTMModel}

        self.results = {}
        self.embedding_matrices = {}

    def create_embedding_matrix(self, emb_name):
        """임베딩 행렬 생성"""
        if emb_name in self.embedding_matrices:
            return self.embedding_matrices[emb_name]

        if emb_name in ['word2vec', 'fasttext']:
            word2idx = self.embedding_obj.models[emb_name].wv.key_to_index
            vocab_size = len(word2idx) + 2
            embedding_dim = self.embedding_obj.vector_size

            embedding_matrix = np.zeros((vocab_size, embedding_dim))
            for word, idx in word2idx.items():
                embedding_matrix[idx + 2] = self.embedding_obj.models[emb_name].wv[word]

        self.embedding_matrices[emb_name] = embedding_matrix
        return embedding_matrix

    def train_epoch(self, model, train_loader, criterion, optimizer):
        """1 에포크 학습"""
        model.train()
        total_loss = 0

        for inputs, labels in train_loader:
            inputs, labels = inputs.to(self.device), labels.to(self.device)

            optimizer.zero_grad()
            outputs = model(inputs)
            loss = criterion(outputs, labels)
            loss.backward()
            optimizer.step()

            total_loss += loss.item()

        return total_loss / len(train_loader)

    def evaluate(self, model, test_loader, return_predictions=False):
        """모델 평가 - 모든 지표 계산"""
        model.eval()
        all_preds = []
        all_labels = []

        with torch.no_grad():
            for inputs, labels in test_loader:
                inputs, labels = inputs.to(self.device), labels.to(self.device)
                outputs = model(inputs)
                _, predicted = torch.max(outputs, 1)

                all_preds.extend(predicted.cpu().numpy())
                all_labels.extend(labels.cpu().numpy())

        # 전체 평가 지표
        accuracy = accuracy_score(all_labels, all_preds)
        precision, recall, f1, _ = precision_recall_fscore_support(
            all_labels, all_preds, average='weighted', zero_division=0
        )

        # 클래스별 평가 지표
        class_precision, class_recall, class_f1, class_support = precision_recall_fscore_support(
            all_labels, all_preds, average=None, zero_division=0
        )

        # Confusion Matrix
        conf_matrix = confusion_matrix(all_labels, all_preds)

        metrics = {
            'accuracy': accuracy * 100,
            'precision': precision * 100,
            'recall': recall * 100,
            'f1': f1 * 100,
            'class_metrics': {
                'precision': class_precision * 100,
                'recall': class_recall * 100,
                'f1': class_f1 * 100,
                'support': class_support
            },
            'confusion_matrix': conf_matrix
        }

        if return_predictions:
            metrics['predictions'] = all_preds
            metrics['labels'] = all_labels

        return metrics

    def train_single_model(self, emb_name, model_name, epochs=10, lr=0.001):
        """단일 조합 학습"""
        print(f"\n{'='*70}")
        print(f"Training: {emb_name.upper()} + {model_name}")
        print(f"{'='*70}")

        embedding_matrix = self.create_embedding_matrix(emb_name)
        train_loader, test_loader = self.dl_manager.get_dataloader(emb_name)

        model_class = self.model_classes[model_name]

        if model_name == 'BiLSTM':
            model = model_class(
                embedding_matrix=embedding_matrix,
                hidden_dim=256,
                output_dim=20,
                n_layers=2,
                dropout=0.3,
                freeze_embeddings=True,
                pooling='last'
            )

        model = model.to(self.device)
        # 손실함수 가중치 제거
        criterion = nn.CrossEntropyLoss()
        optimizer = optim.Adam(model.parameters(), lr=lr)

        best_f1 = 0
        best_metrics = None

        for epoch in range(epochs):
            train_loss = self.train_epoch(model, train_loader, criterion, optimizer)
            metrics = self.evaluate(model, test_loader, return_predictions=True)

            if metrics['f1'] > best_f1:
                best_f1 = metrics['f1']
                best_metrics = metrics.copy()

            if (epoch + 1) % 2 == 0:
                print(f"Epoch {epoch+1}/{epochs} - Loss: {train_loss:.4f}, "
                      f"F1: {metrics['f1']:.2f}%, Acc: {metrics['accuracy']:.2f}%")

        total_params = sum(p.numel() for p in model.parameters())

        key = f"{emb_name}_{model_name}"
        self.results[key] = {
            'embedding': emb_name,
            'model': model_name,
            'metrics': best_metrics,
            'params': total_params
        }

        print(f"✓ Best F1: {best_f1:.2f}%")
        return best_metrics

    def train_all_combinations(self, epochs=10, lr=0.001):
        """2가지 임베딩에 대한 BiLSTM 학습"""
        print("\n" + "="*70)
        print("Training All Embedding of BiLSTM")
        print("="*70)

        for emb_name in self.embedding_names:
            for model_name in self.model_classes.keys():
                self.train_single_model(emb_name, model_name, epochs, lr)

        print("\n✓ All combinations trained!")

    def get_results_dataframe(self):
        """결과를 DataFrame으로 반환"""
        data = []
        for key, result in self.results.items():
            row = {
                'Embedding': result['embedding'].upper(),
                'Model': result['model'],
                'Accuracy': result['metrics']['accuracy'],
                'Precision': result['metrics']['precision'],
                'Recall': result['metrics']['recall'],
                'F1-Score': result['metrics']['f1'],
                'Parameters': result['params']
            }
            data.append(row)

        df = pd.DataFrame(data)
        df = df.sort_values('F1-Score', ascending=False)
        return df

    def print_class_report(self, combination_key=None):
        """특정 조합의 클래스별 상세 리포트"""
        if combination_key is None:
            df = self.get_results_dataframe()
            best = df.iloc[0]
            combination_key = f"{best['Embedding'].lower()}_{best['Model']}"

        result = self.results[combination_key]
        class_metrics = result['metrics']['class_metrics']

        print("\n" + "="*100)
        print(f"CLASS-WISE PERFORMANCE: {result['embedding'].upper()} + {result['model']}")
        print("="*100)
        print(f"{'Class':<25}{'Precision':<12}{'Recall':<12}{'F1-Score':<12}{'Support':<12}")
        print("-"*100)

        for i, name in enumerate(self.class_names):
            print(f"{name:<25}{class_metrics['precision'][i]:<12.2f}"
                  f"{class_metrics['recall'][i]:<12.2f}"
                  f"{class_metrics['f1'][i]:<12.2f}"
                  f"{int(class_metrics['support'][i]):<12}")

        print("-"*100)
        print(f"{'OVERALL':<25}{result['metrics']['precision']:<12.2f}"
              f"{result['metrics']['recall']:<12.2f}"
              f"{result['metrics']['f1']:<12.2f}"
              f"{int(class_metrics['support'].sum()):<12}")
        print("="*100)

    def print_best_combination(self):
        """최고 성능 조합 출력"""
        df = self.get_results_dataframe()
        best = df.iloc[0]

        print("\n" + "="*70)
        print("BEST COMBINATION (Based on F1-Score)")
        print("="*70)
        print(f"\n🏆 Best Embedding: {best['Embedding']}")
        print(f"🏆 Best Model:     {best['Model']}")
        print(f"\n📊 Performance Metrics:")
        print(f"   • Accuracy:  {best['Accuracy']:.2f}%")
        print(f"   • Precision: {best['Precision']:.2f}%")
        print(f"   • Recall:    {best['Recall']:.2f}%")
        print(f"   • F1-Score:  {best['F1-Score']:.2f}%")
        print(f"\n⚙️  Model Parameters: {best['Parameters']:,}")
        print("="*70)

    def print_full_results_table(self):
        """전체 결과 테이블 출력"""
        df = self.get_results_dataframe()

        print("\n" + "="*100)
        print("FULL RESULTS TABLE (Sorted by F1-Score)")
        print("="*100)
        print(f"{'Rank':<6}{'Embedding':<12}{'Model':<10}{'Acc(%)':<10}{'Prec(%)':<10}"
              f"{'Rec(%)':<10}{'F1(%)':<10}{'Params':<15}")
        print("-"*100)

        for idx, (_, row) in enumerate(df.iterrows(), 1):
            print(f"{idx:<6}{row['Embedding']:<12}{row['Model']:<10}"
                  f"{row['Accuracy']:<10.2f}{row['Precision']:<10.2f}"
                  f"{row['Recall']:<10.2f}{row['F1-Score']:<10.2f}"
                  f"{row['Parameters']:<15,}")

        print("="*100)

    def save_results(self, filepath='./results.csv'):
        """결과를 CSV로 저장"""
        df = self.get_results_dataframe()
        df.to_csv(filepath, index=False)
        print(f"\n✓ Results saved to {filepath}")
````

````python
def clean_text_with_nltk(text):
    # 소문자 변환
    text = text.lower()

    # 특수문자 제거 (선택사항 - tokenize가 알아서 처리하기도 함)
    text = re.sub(r'[^a-zA-Z\s]', '', text)

    # NLTK word_tokenize 사용
    tokens = word_tokenize(text)

    # 불용어 제거
    stop_words = set(stopwords.words('english'))
    tokens = [word for word in tokens if word not in stop_words]

    return ' '.join(tokens)
````

````python
# 문장 길이 기준으로 데이터 필터링
def filter_by_length(texts, labels, min_len=10, max_len=500):

    from nltk.tokenize import word_tokenize

    filtered_texts = []
    filtered_labels = []

    removed_too_short = 0
    removed_too_long = 0

    for text, label in zip(texts, labels):
        # 단어 수 계산
        word_count = len(word_tokenize(text))

        # 길이 체크
        if word_count < min_len:
            removed_too_short += 1
            continue
        elif word_count > max_len:
            removed_too_long += 1
            continue

        filtered_texts.append(text)
        filtered_labels.append(label)

    print(f"\n{'='*70}")
    print("문장 길이 필터링 결과")
    print(f"{'='*70}")
    print(f"원본 데이터:           {len(texts):,}개")
    print(f"너무 짧음 제거 (<{min_len}):  {removed_too_short:,}개")
    print(f"너무 김 제거 (>{max_len}):    {removed_too_long:,}개")
    print(f"최종 데이터:           {len(filtered_texts):,}개")
    print(f"유지 비율:             {len(filtered_texts)/len(texts)*100:.1f}%")
    print(f"{'='*70}\n")

    return filtered_texts, np.array(filtered_labels)
````

### 텍스트 임베딩

````python
class MultiEmbedding:
    """Word2Vec, FastText를 한 번에 학습/변환하는 통합 클래스"""

    def __init__(self,
                 vector_size=100,
                 window=5,
                 min_count=5,
                 sg=1,
                 epochs=10,
                 workers=4,
                 max_features=10000,
                 min_df=5,
                 seed=156,
                 embeddings=['word2vec', 'fasttext']):

        self.vector_size = vector_size
        self.window = window
        self.min_count = min_count
        self.sg = sg
        self.epochs = epochs
        self.workers = workers
        self.max_features = max_features
        self.min_df = min_df
        self.seed = seed
        self.embeddings = embeddings

        # 모델 저장
        self.models = {}
        self.is_trained = False

    def fit(self, X_train):
        """모든 임베딩 모델 학습"""
        print("\n" + "="*70)
        print("통합 임베딩 학습 시작")
        print("="*70)

        if 'word2vec' in self.embeddings:
            self._train_word2vec(X_train)

        if 'fasttext' in self.embeddings:
            self._train_fasttext(X_train)

        self.is_trained = True
        print("\n" + "="*70)
        print("모든 임베딩 학습 완료!")
        print("="*70 + "\n")

        return self

    def _train_word2vec(self, X_train):
        """Word2Vec 학습"""
        print("\n[1/3] Word2Vec 학습 중...")
        print("-" * 70)

        train_sentences = [text.split() for text in X_train]

        self.models['word2vec'] = Word2Vec(
            sentences=train_sentences,
            vector_size=self.vector_size,
            window=self.window,
            min_count=self.min_count,
            workers=self.workers,
            sg=self.sg,
            epochs=self.epochs,
            seed=self.seed
        )

        print(f"✓ Word2Vec Vocabulary 크기: {len(self.models['word2vec'].wv)}")

    def _train_fasttext(self, X_train):
        """FastText 학습"""
        print("\n[2/3] FastText 학습 중...")
        print("-" * 70)

        train_sentences = [text.split() for text in X_train]

        self.models['fasttext'] = FastText(
            sentences=train_sentences,
            vector_size=self.vector_size,
            window=self.window,
            min_count=self.min_count,
            workers=self.workers,
            sg=self.sg,
            epochs=self.epochs,
            seed=self.seed
        )

        print(f"✓ FastText Vocabulary 크기: {len(self.models['fasttext'].wv)}")

    def transform(self, X):
        """모든 임베딩으로 문서 변환"""
        if not self.is_trained:
            raise ValueError("모델이 학습되지 않았습니다. fit()을 먼저 호출하세요.")

        print("\n" + "="*70)
        print("데이터 벡터화 진행")
        print("="*70)

        results = {}

        if 'word2vec' in self.embeddings:
            results['word2vec'] = self._transform_word2vec(X)

        if 'fasttext' in self.embeddings:
            results['fasttext'] = self._transform_fasttext(X)

        print("\n✓ 모든 벡터화 완료!\n")

        return results

    def _transform_word2vec(self, X):
        """Word2Vec 변환"""
        print("\n[Word2Vec 벡터화]")
        model = self.models['word2vec']

        def document_vector(doc):
            if isinstance(doc, str):
                doc = doc.split()
            vectors = [model.wv[word] for word in doc if word in model.wv]
            return np.mean(vectors, axis=0) if vectors else np.zeros(self.vector_size)

        vectors = np.array([document_vector(doc) for doc in X])
        print(f"  Shape: {vectors.shape}")
        return vectors

    def _transform_fasttext(self, X):
        """FastText 변환"""
        print("\n[FastText 벡터화]")
        model = self.models['fasttext']

        def document_vector(doc):
            if isinstance(doc, str):
                doc = doc.split()
            vectors = [model.wv[word] for word in doc]
            return np.mean(vectors, axis=0) if vectors else np.zeros(self.vector_size)

        vectors = np.array([document_vector(doc) for doc in X])
        print(f"  Shape: {vectors.shape}")
        return vectors

    def fit_transform(self, X_train):
        """학습 + 변환"""
        self.fit(X_train)
        return self.transform(X_train)

    def save_models(self, prefix='./'):
        """모든 모델 저장"""
        print("\n" + "="*70)
        print("모델 저장 중")
        print("="*70)

        if 'word2vec' in self.models:
            filepath = f'{prefix}word2vec_model.model'
            self.models['word2vec'].save(filepath)
            print(f"✓ Word2Vec 저장: {filepath}")

        if 'fasttext' in self.models:
            filepath = f'{prefix}fasttext_model.model'
            self.models['fasttext'].save(filepath)
            print(f"✓ FastText 저장: {filepath}")

        print("✓ 모든 모델 저장 완료!\n")

    def get_vocab_info(self):
        """각 임베딩의 vocabulary 정보 출력"""
        print("\n" + "="*70)
        print("Vocabulary 정보")
        print("="*70)

        if 'word2vec' in self.models:
            print(f"Word2Vec: {len(self.models['word2vec'].wv)} 단어")

        if 'fasttext' in self.models:
            print(f"FastText:  {len(self.models['fasttext'].wv)} 단어")

        print()
````

### 데이터셋 클래스, 데이터 로더 클래스

````python
class TextDataset(Dataset):
    def __init__(self, texts, labels, word2idx, max_len,
                 pad_idx=0, unk_idx=1, lowercase=True, add_offset=True):
        self.texts = texts
        self.labels = labels
        self.lowercase = lowercase
        self.max_len = max_len
        self.pad_idx = pad_idx
        self.unk_idx = unk_idx
        self.add_offset = add_offset

        # ✅ word2idx를 미리 +2 오프셋으로 변환
        if add_offset:
            self.word2idx = {word: idx + 2 for word, idx in word2idx.items()}
        else:
            self.word2idx = word2idx

    def __len__(self):
        return len(self.texts)

    def __getitem__(self, idx):
        text = self.texts[idx]

        if self.lowercase:
            text = text.lower()

        tokens = word_tokenize(text)
        encoded = [self.word2idx.get(word, self.unk_idx) for word in tokens]

        if len(encoded) < self.max_len:
            encoded += [self.pad_idx] * (self.max_len - len(encoded))
        else:
            encoded = encoded[:self.max_len]

        encoded_tensor = torch.tensor(encoded, dtype=torch.long)
        label_tensor = torch.tensor(self.labels[idx], dtype=torch.long)

        return encoded_tensor, label_tensor
````

````python
class EmbeddingDataLoaderManager:
    """3가지 임베딩의 DataLoader를 관리하는 클래스"""

    def __init__(self, embedding, X_train, y_train, X_test, y_test,
                 max_len=256, batch_size=64, pad_idx=0, unk_idx=1):

        self.embedding = embedding
        self.X_train = X_train
        self.y_train = y_train
        self.X_test = X_test
        self.y_test = y_test
        self.max_len = max_len
        self.batch_size = batch_size
        self.pad_idx = pad_idx
        self.unk_idx = unk_idx

        self.dataloaders = {}
        self._create_all_dataloaders()

    def _create_all_dataloaders(self):
        """모든 임베딩에 대한 DataLoader 생성"""
        print("\n" + "="*70)
        print("DataLoader 생성 중...")
        print("="*70)

        # Word2Vec
        if 'word2vec' in self.embedding.models:
            self.dataloaders['word2vec'] = self._create_dataloader('word2vec')

        # FastText
        if 'fasttext' in self.embedding.models:
            self.dataloaders['fasttext'] = self._create_dataloader('fasttext')

        print("\n✓ 모든 DataLoader 생성 완료!\n")

    def _create_dataloader(self, emb_name):
        """특정 임베딩에 대한 DataLoader 생성"""
        # word2idx 추출
        if emb_name in ['word2vec', 'fasttext']:
            word2idx = self.embedding.models[emb_name].wv.key_to_index
        else:
            print("없는 임베딩입니다.")

        # Dataset 생성 (✅ add_offset=True 추가)
        train_dataset = TextDataset(
            self.X_train, self.y_train, word2idx,
            self.max_len, self.pad_idx, self.unk_idx,
            add_offset=True  # ✅ 추가!
        )
        test_dataset = TextDataset(
            self.X_test, self.y_test, word2idx,
            self.max_len, self.pad_idx, self.unk_idx,
            add_offset=True  # ✅ 추가!
        )

        # DataLoader 생성
        train_loader = DataLoader(train_dataset, batch_size=self.batch_size, shuffle=True)
        test_loader = DataLoader(test_dataset, batch_size=self.batch_size, shuffle=False)

        print(f"  ✓ {emb_name.upper():10s} (vocab: {len(word2idx):5d} 단어)")

        return train_loader, test_loader

    def get_dataloader(self, emb_name):
        """특정 임베딩의 DataLoader 가져오기"""
        if emb_name not in self.dataloaders:
            raise ValueError(f"{emb_name}에 대한 DataLoader가 없습니다.")
        return self.dataloaders[emb_name]

    def get_vocab_size(self, emb_name):
        """특정 임베딩의 vocabulary 크기"""
        if emb_name in ['word2vec', 'fasttext']:
            return len(self.embedding.models[emb_name].wv.key_to_index)
        else:
            print("없는 임베딩입니다.")

    def show_info(self):
        """DataLoader 정보 출력"""
        print("\n" + "="*70)
        print("DataLoader 정보")
        print("="*70)

        for emb_name in self.dataloaders.keys():
            train_loader, test_loader = self.dataloaders[emb_name]
            vocab_size = self.get_vocab_size(emb_name)

            print(f"\n[{emb_name.upper()}]")
            print(f"  Vocabulary 크기: {vocab_size}")
            print(f"  Train batches: {len(train_loader)}")
            print(f"  Test batches:  {len(test_loader)}")
            print(f"  Batch size:    {self.batch_size}")
            print(f"  Max length:    {self.max_len}")
````

### 모델 클래스(양방향 LSTM)

````python
class BiLSTMModel(nn.Module):
    def __init__(self, embedding_matrix, hidden_dim, output_dim,
                 n_layers=2, dropout=0.3, freeze_embeddings=False, pooling='last'):
        super(BiLSTMModel, self).__init__()

        self.hidden_dim = hidden_dim
        self.n_layers = n_layers
        self.pooling = pooling

        # 임베딩 레이어
        self.embedding = nn.Embedding.from_pretrained(
            torch.tensor(embedding_matrix, dtype=torch.float),
            freeze=freeze_embeddings
        )

        embedding_dim = embedding_matrix.shape[1]

        # 양방향 LSTM
        self.lstm = nn.LSTM(
            embedding_dim,
            hidden_dim,
            num_layers=n_layers,
            dropout=dropout if n_layers > 1 else 0,
            batch_first=True,
            bidirectional=True  # ✅ 양방향!
        )

        # 양방향이므로 출력 차원 2배
        lstm_output_dim = hidden_dim * 2

        # 드롭아웃
        self.dropout = nn.Dropout(dropout)

        # Fully Connected Layers
        self.fc1 = nn.Linear(lstm_output_dim, 128)  # ✅ 2배 차원
        self.fc2 = nn.Linear(128, 64)
        self.fc3 = nn.Linear(64, output_dim)

    def forward(self, x):
        # 임베딩
        embedded = self.embedding(x)

        # 양방향 LSTM
        lstm_out, (hidden, cell) = self.lstm(embedded)
        # lstm_out: (batch, seq_len, hidden_dim * 2)
        # hidden: (n_layers * 2, batch, hidden_dim)

        # Pooling
        if self.pooling == 'last':
            # Forward와 Backward의 마지막 hidden state concat
            # hidden[-2]: forward의 마지막 레이어
            # hidden[-1]: backward의 마지막 레이어
            output = torch.cat((hidden[-2], hidden[-1]), dim=1)

        elif self.pooling == 'mean':
            # LSTM 출력의 평균
            output = torch.mean(lstm_out, dim=1)

        elif self.pooling == 'max':
            # LSTM 출력의 최댓값
            output = torch.max(lstm_out, dim=1)[0]
        else:
            raise ValueError(f"Unknown pooling: {self.pooling}")

        # FC Layers
        output = self.dropout(output)
        output = F.relu(self.fc1(output))
        output = self.dropout(output)
        output = F.relu(self.fc2(output))
        output = self.dropout(output)
        output = self.fc3(output)

        return output
````

### 모델 학습 및 평가

````python
class EmbeddingModelEvaluator:
    """3가지 임베딩 × 3가지 모델 = 9개 조합 평가 시스템"""

    def __init__(self, embedding_obj, dl_manager, class_names, device='cuda'):

        self.embedding_obj = embedding_obj
        self.dl_manager = dl_manager
        self.class_names = class_names
        self.device = torch.device(device if torch.cuda.is_available() else 'cpu')

        self.embedding_names = ['word2vec', 'fasttext']
        self.model_classes = {'BiLSTM': BiLSTMModel}

        self.results = {}
        self.embedding_matrices = {}

    def create_embedding_matrix(self, emb_name):
        """임베딩 행렬 생성"""
        if emb_name in self.embedding_matrices:
            return self.embedding_matrices[emb_name]

        if emb_name in ['word2vec', 'fasttext']:
            word2idx = self.embedding_obj.models[emb_name].wv.key_to_index
            vocab_size = len(word2idx) + 2
            embedding_dim = self.embedding_obj.vector_size

            embedding_matrix = np.zeros((vocab_size, embedding_dim))
            for word, idx in word2idx.items():
                embedding_matrix[idx + 2] = self.embedding_obj.models[emb_name].wv[word]

        self.embedding_matrices[emb_name] = embedding_matrix
        return embedding_matrix

    def train_epoch(self, model, train_loader, criterion, optimizer):
        """1 에포크 학습"""
        model.train()
        total_loss = 0

        for inputs, labels in train_loader:
            inputs, labels = inputs.to(self.device), labels.to(self.device)

            optimizer.zero_grad()
            outputs = model(inputs)
            loss = criterion(outputs, labels)
            loss.backward()
            optimizer.step()

            total_loss += loss.item()

        return total_loss / len(train_loader)

    def evaluate(self, model, test_loader, return_predictions=False):
        """모델 평가 - 모든 지표 계산"""
        model.eval()
        all_preds = []
        all_labels = []

        with torch.no_grad():
            for inputs, labels in test_loader:
                inputs, labels = inputs.to(self.device), labels.to(self.device)
                outputs = model(inputs)
                _, predicted = torch.max(outputs, 1)

                all_preds.extend(predicted.cpu().numpy())
                all_labels.extend(labels.cpu().numpy())

        # 전체 평가 지표
        accuracy = accuracy_score(all_labels, all_preds)
        precision, recall, f1, _ = precision_recall_fscore_support(
            all_labels, all_preds, average='weighted', zero_division=0
        )

        # 클래스별 평가 지표
        class_precision, class_recall, class_f1, class_support = precision_recall_fscore_support(
            all_labels, all_preds, average=None, zero_division=0
        )

        # Confusion Matrix
        conf_matrix = confusion_matrix(all_labels, all_preds)

        metrics = {
            'accuracy': accuracy * 100,
            'precision': precision * 100,
            'recall': recall * 100,
            'f1': f1 * 100,
            'class_metrics': {
                'precision': class_precision * 100,
                'recall': class_recall * 100,
                'f1': class_f1 * 100,
                'support': class_support
            },
            'confusion_matrix': conf_matrix
        }

        if return_predictions:
            metrics['predictions'] = all_preds
            metrics['labels'] = all_labels

        return metrics

    def train_single_model(self, emb_name, model_name, epochs=10, lr=0.001):
        """단일 조합 학습"""
        print(f"\n{'='*70}")
        print(f"Training: {emb_name.upper()} + {model_name}")
        print(f"{'='*70}")

        embedding_matrix = self.create_embedding_matrix(emb_name)
        train_loader, test_loader = self.dl_manager.get_dataloader(emb_name)

        model_class = self.model_classes[model_name]

        if model_name == 'BiLSTM':
            model = model_class(
                embedding_matrix=embedding_matrix,
                hidden_dim=256,
                output_dim=20,
                n_layers=2,
                dropout=0.3,
                freeze_embeddings=True,
                pooling='last'
            )

        model = model.to(self.device)
        # 손실함수 가중치 제거
        criterion = nn.CrossEntropyLoss()
        optimizer = optim.Adam(model.parameters(), lr=lr)

        best_f1 = 0
        best_metrics = None

        for epoch in range(epochs):
            train_loss = self.train_epoch(model, train_loader, criterion, optimizer)
            metrics = self.evaluate(model, test_loader, return_predictions=True)

            if metrics['f1'] > best_f1:
                best_f1 = metrics['f1']
                best_metrics = metrics.copy()

            if (epoch + 1) % 2 == 0:
                print(f"Epoch {epoch+1}/{epochs} - Loss: {train_loss:.4f}, "
                      f"F1: {metrics['f1']:.2f}%, Acc: {metrics['accuracy']:.2f}%")

        total_params = sum(p.numel() for p in model.parameters())

        key = f"{emb_name}_{model_name}"
        self.results[key] = {
            'embedding': emb_name,
            'model': model_name,
            'metrics': best_metrics,
            'params': total_params
        }

        print(f"✓ Best F1: {best_f1:.2f}%")
        return best_metrics

    def train_all_combinations(self, epochs=10, lr=0.001):
        """2가지 임베딩에 대한 BiLSTM 학습"""
        print("\n" + "="*70)
        print("Training All Embedding of BiLSTM")
        print("="*70)

        for emb_name in self.embedding_names:
            for model_name in self.model_classes.keys():
                self.train_single_model(emb_name, model_name, epochs, lr)

        print("\n✓ All combinations trained!")

    def get_results_dataframe(self):
        """결과를 DataFrame으로 반환"""
        data = []
        for key, result in self.results.items():
            row = {
                'Embedding': result['embedding'].upper(),
                'Model': result['model'],
                'Accuracy': result['metrics']['accuracy'],
                'Precision': result['metrics']['precision'],
                'Recall': result['metrics']['recall'],
                'F1-Score': result['metrics']['f1'],
                'Parameters': result['params']
            }
            data.append(row)

        df = pd.DataFrame(data)
        df = df.sort_values('F1-Score', ascending=False)
        return df

    def print_class_report(self, combination_key=None):
        """특정 조합의 클래스별 상세 리포트"""
        if combination_key is None:
            df = self.get_results_dataframe()
            best = df.iloc[0]
            combination_key = f"{best['Embedding'].lower()}_{best['Model']}"

        result = self.results[combination_key]
        class_metrics = result['metrics']['class_metrics']

        print("\n" + "="*100)
        print(f"CLASS-WISE PERFORMANCE: {result['embedding'].upper()} + {result['model']}")
        print("="*100)
        print(f"{'Class':<25}{'Precision':<12}{'Recall':<12}{'F1-Score':<12}{'Support':<12}")
        print("-"*100)

        for i, name in enumerate(self.class_names):
            print(f"{name:<25}{class_metrics['precision'][i]:<12.2f}"
                  f"{class_metrics['recall'][i]:<12.2f}"
                  f"{class_metrics['f1'][i]:<12.2f}"
                  f"{int(class_metrics['support'][i]):<12}")

        print("-"*100)
        print(f"{'OVERALL':<25}{result['metrics']['precision']:<12.2f}"
              f"{result['metrics']['recall']:<12.2f}"
              f"{result['metrics']['f1']:<12.2f}"
              f"{int(class_metrics['support'].sum()):<12}")
        print("="*100)

    def print_best_combination(self):
        """최고 성능 조합 출력"""
        df = self.get_results_dataframe()
        best = df.iloc[0]

        print("\n" + "="*70)
        print("BEST COMBINATION (Based on F1-Score)")
        print("="*70)
        print(f"\n🏆 Best Embedding: {best['Embedding']}")
        print(f"🏆 Best Model:     {best['Model']}")
        print(f"\n📊 Performance Metrics:")
        print(f"   • Accuracy:  {best['Accuracy']:.2f}%")
        print(f"   • Precision: {best['Precision']:.2f}%")
        print(f"   • Recall:    {best['Recall']:.2f}%")
        print(f"   • F1-Score:  {best['F1-Score']:.2f}%")
        print(f"\n⚙️  Model Parameters: {best['Parameters']:,}")
        print("="*70)

    def print_full_results_table(self):
        """전체 결과 테이블 출력"""
        df = self.get_results_dataframe()

        print("\n" + "="*100)
        print("FULL RESULTS TABLE (Sorted by F1-Score)")
        print("="*100)
        print(f"{'Rank':<6}{'Embedding':<12}{'Model':<10}{'Acc(%)':<10}{'Prec(%)':<10}"
              f"{'Rec(%)':<10}{'F1(%)':<10}{'Params':<15}")
        print("-"*100)

        for idx, (_, row) in enumerate(df.iterrows(), 1):
            print(f"{idx:<6}{row['Embedding']:<12}{row['Model']:<10}"
                  f"{row['Accuracy']:<10.2f}{row['Precision']:<10.2f}"
                  f"{row['Recall']:<10.2f}{row['F1-Score']:<10.2f}"
                  f"{row['Parameters']:<15,}")

        print("="*100)

    def save_results(self, filepath='./results.csv'):
        """결과를 CSV로 저장"""
        df = self.get_results_dataframe()
        df.to_csv(filepath, index=False)
        print(f"\n✓ Results saved to {filepath}")
````

````python
# 문장 길이 기준으로 데이터 필터링
def filter_by_length(texts, labels, min_len=10, max_len=500):

    from nltk.tokenize import word_tokenize

    filtered_texts = []
    filtered_labels = []

    removed_too_short = 0
    removed_too_long = 0

    for text, label in zip(texts, labels):
        # 단어 수 계산
        word_count = len(word_tokenize(text))

        # 길이 체크
        if word_count < min_len:
            removed_too_short += 1
            continue
        elif word_count > max_len:
            removed_too_long += 1
            continue

        filtered_texts.append(text)
        filtered_labels.append(label)

    print(f"\n{'='*70}")
    print("문장 길이 필터링 결과")
    print(f"{'='*70}")
    print(f"원본 데이터:           {len(texts):,}개")
    print(f"너무 짧음 제거 (<{min_len}):  {removed_too_short:,}개")
    print(f"너무 김 제거 (>{max_len}):    {removed_too_long:,}개")
    print(f"최종 데이터:           {len(filtered_texts):,}개")
    print(f"유지 비율:             {len(filtered_texts)/len(texts)*100:.1f}%")
    print(f"{'='*70}\n")

    return filtered_texts, np.array(filtered_labels)
````

## 텍스트 임베딩

**임베딩 종류**
1. Word2Vec
2. fastText
3. Clove

3가지 모두 실습해야해서 통합 임베딩 파이프라인을 작성했다.

````python
class MultiEmbedding:
    """Word2Vec, FastText, GloVe를 한 번에 학습/변환하는 통합 클래스"""

    def __init__(self,
                 vector_size=100,
                 window=5,
                 min_count=5,
                 sg=1,
                 epochs=10,
                 workers=4,
                 max_features=10000,
                 min_df=5,
                 glove_max_iter=10,
                 glove_lr=0.05,
                 seed=156,
                 embeddings=['word2vec', 'fasttext', 'glove']):

        self.vector_size = vector_size
        self.window = window
        self.min_count = min_count
        self.sg = sg
        self.epochs = epochs
        self.workers = workers
        self.max_features = max_features
        self.min_df = min_df
        self.glove_max_iter = glove_max_iter
        self.glove_lr = glove_lr
        self.seed = seed
        self.embeddings = embeddings

        # 모델 저장
        self.models = {}
        self.is_trained = False

    def fit(self, X_train):
        """모든 임베딩 모델 학습"""
        print("\n" + "="*70)
        print("통합 임베딩 학습 시작")
        print("="*70)

        if 'word2vec' in self.embeddings:
            self._train_word2vec(X_train)

        if 'fasttext' in self.embeddings:
            self._train_fasttext(X_train)

        if 'glove' in self.embeddings:
            self._train_glove(X_train)

        self.is_trained = True
        print("\n" + "="*70)
        print("모든 임베딩 학습 완료!")
        print("="*70 + "\n")

        return self

    def _train_word2vec(self, X_train):
        """Word2Vec 학습"""
        print("\n[1/3] Word2Vec 학습 중...")
        print("-" * 70)

        train_sentences = [text.split() for text in X_train]

        self.models['word2vec'] = Word2Vec(
            sentences=train_sentences,
            vector_size=self.vector_size,
            window=self.window,
            min_count=self.min_count,
            workers=self.workers,
            sg=self.sg,
            epochs=self.epochs,
            seed=self.seed
        )

        print(f"✓ Word2Vec Vocabulary 크기: {len(self.models['word2vec'].wv)}")

    def _train_fasttext(self, X_train):
        """FastText 학습"""
        print("\n[2/3] FastText 학습 중...")
        print("-" * 70)

        train_sentences = [text.split() for text in X_train]

        self.models['fasttext'] = FastText(
            sentences=train_sentences,
            vector_size=self.vector_size,
            window=self.window,
            min_count=self.min_count,
            workers=self.workers,
            sg=self.sg,
            epochs=self.epochs,
            seed=self.seed
        )

        print(f"✓ FastText Vocabulary 크기: {len(self.models['fasttext'].wv)}")

    def _train_glove(self, X_train):
        """GloVe 학습"""
        print("\n[3/3] GloVe 학습 중...")
        print("-" * 70)

        # 어휘 사전 구축
        vectorizer = CountVectorizer(
            max_features=self.max_features,
            min_df=self.min_df
        )
        vectorizer.fit(X_train)
        vocab = vectorizer.get_feature_names_out()
        word_to_idx = {word: idx for idx, word in enumerate(vocab)}
        vocab_size = len(vocab)

        print(f"  어휘 사전 구축 완료: {vocab_size}개 단어")

        # 동시출현 행렬 생성
        print("  동시출현 행렬 생성 중...")
        cooccurrence_matrix = np.zeros((vocab_size, vocab_size))

        for doc in X_train:
            words = doc.split()
            words = [w for w in words if w in word_to_idx]

            for i, word in enumerate(words):
                word_idx = word_to_idx[word]
                start = max(0, i - self.window)
                end = min(len(words), i + self.window + 1)

                for j in range(start, end):
                    if i != j:
                        context_word = words[j]
                        context_idx = word_to_idx[context_word]
                        distance = abs(i - j)
                        cooccurrence_matrix[word_idx, context_idx] += 1.0 / distance

        # GloVe 학습
        print("  GloVe 모델 학습 중...")
        glove_model = GloVe(
            n=self.vector_size,
            max_iter=self.glove_max_iter,
            learning_rate=self.glove_lr,
            display_progress=False
        )

        embeddings = glove_model.fit(cooccurrence_matrix)

        # 저장
        self.models['glove'] = {
            'word_to_vec': {word: embeddings[idx] for word, idx in word_to_idx.items()},
            'vocab': vocab,
            'word_to_idx': word_to_idx
        }

        print(f"✓ GloVe Vocabulary 크기: {vocab_size}")

    def transform(self, X):
        """모든 임베딩으로 문서 변환"""
        if not self.is_trained:
            raise ValueError("모델이 학습되지 않았습니다. fit()을 먼저 호출하세요.")

        print("\n" + "="*70)
        print("데이터 벡터화 진행")
        print("="*70)

        results = {}

        if 'word2vec' in self.embeddings:
            results['word2vec'] = self._transform_word2vec(X)

        if 'fasttext' in self.embeddings:
            results['fasttext'] = self._transform_fasttext(X)

        if 'glove' in self.embeddings:
            results['glove'] = self._transform_glove(X)

        print("\n✓ 모든 벡터화 완료!\n")

        return results

    def _transform_word2vec(self, X):
        """Word2Vec 변환"""
        print("\n[Word2Vec 벡터화]")
        model = self.models['word2vec']

        def document_vector(doc):
            if isinstance(doc, str):
                doc = doc.split()
            vectors = [model.wv[word] for word in doc if word in model.wv]
            return np.mean(vectors, axis=0) if vectors else np.zeros(self.vector_size)

        vectors = np.array([document_vector(doc) for doc in X])
        print(f"  Shape: {vectors.shape}")
        return vectors

    def _transform_fasttext(self, X):
        """FastText 변환"""
        print("\n[FastText 벡터화]")
        model = self.models['fasttext']

        def document_vector(doc):
            if isinstance(doc, str):
                doc = doc.split()
            vectors = [model.wv[word] for word in doc]
            return np.mean(vectors, axis=0) if vectors else np.zeros(self.vector_size)

        vectors = np.array([document_vector(doc) for doc in X])
        print(f"  Shape: {vectors.shape}")
        return vectors

    def _transform_glove(self, X):
        """GloVe 변환"""
        print("\n[GloVe 벡터화]")
        word_to_vec = self.models['glove']['word_to_vec']

        def document_vector(doc):
            if isinstance(doc, str):
                doc = doc.split()
            vectors = [word_to_vec[word] for word in doc if word in word_to_vec]
            return np.mean(vectors, axis=0) if vectors else np.zeros(self.vector_size)

        vectors = np.array([document_vector(doc) for doc in X])
        print(f"  Shape: {vectors.shape}")
        return vectors

    def fit_transform(self, X_train):
        """학습 + 변환"""
        self.fit(X_train)
        return self.transform(X_train)

    def save_models(self, prefix='./'):
        """모든 모델 저장"""
        print("\n" + "="*70)
        print("모델 저장 중")
        print("="*70)

        if 'word2vec' in self.models:
            filepath = f'{prefix}word2vec_model.model'
            self.models['word2vec'].save(filepath)
            print(f"✓ Word2Vec 저장: {filepath}")

        if 'fasttext' in self.models:
            filepath = f'{prefix}fasttext_model.model'
            self.models['fasttext'].save(filepath)
            print(f"✓ FastText 저장: {filepath}")

        if 'glove' in self.models:
            filepath = f'{prefix}glove_model.npz'
            np.savez(
                filepath,
                embeddings=np.array([self.models['glove']['word_to_vec'][word]
                                    for word in self.models['glove']['vocab']]),
                vocab=self.models['glove']['vocab'],
                vector_size=self.vector_size
            )
            print(f"✓ GloVe 저장: {filepath}")

        print("✓ 모든 모델 저장 완료!\n")

    def get_vocab_info(self):
        """각 임베딩의 vocabulary 정보 출력"""
        print("\n" + "="*70)
        print("Vocabulary 정보")
        print("="*70)

        if 'word2vec' in self.models:
            print(f"Word2Vec: {len(self.models['word2vec'].wv)} 단어")

        if 'fasttext' in self.models:
            print(f"FastText:  {len(self.models['fasttext'].wv)} 단어")

        if 'glove' in self.models:
            print(f"GloVe:     {len(self.models['glove']['vocab'])} 단어")

        print()
````

## 데이터셋 및 로더 생성

````python
class TextDataset(Dataset):
    def __init__(self, texts, labels, word2idx, max_len,
                 pad_idx=0, unk_idx=1, lowercase=True, add_offset=True):
        self.texts = texts
        self.labels = labels
        self.lowercase = lowercase
        self.max_len = max_len
        self.pad_idx = pad_idx
        self.unk_idx = unk_idx
        self.add_offset = add_offset

        # ✅ word2idx를 미리 +2 오프셋으로 변환
        if add_offset:
            self.word2idx = {word: idx + 2 for word, idx in word2idx.items()}
        else:
            self.word2idx = word2idx

    def __len__(self):
        return len(self.texts)

    def __getitem__(self, idx):
        text = self.texts[idx]

        if self.lowercase:
            text = text.lower()

        tokens = word_tokenize(text)
        encoded = [self.word2idx.get(word, self.unk_idx) for word in tokens]

        if len(encoded) < self.max_len:
            encoded += [self.pad_idx] * (self.max_len - len(encoded))
        else:
            encoded = encoded[:self.max_len]

        encoded_tensor = torch.tensor(encoded, dtype=torch.long)
        label_tensor = torch.tensor(self.labels[idx], dtype=torch.long)

        return encoded_tensor, label_tensor
````

````python
class EmbeddingDataLoaderManager:
    """3가지 임베딩의 DataLoader를 관리하는 클래스"""

    def __init__(self, embedding, X_train, y_train, X_test, y_test,
                 max_len=256, batch_size=64, pad_idx=0, unk_idx=1):
        """
        Parameters:
        -----------
        embedding : MultiEmbedding
            학습된 MultiEmbedding 객체
        X_train, y_train : array-like
            학습 데이터
        X_test, y_test : array-like
            테스트 데이터
        max_len : int
            최대 시퀀스 길이
        batch_size : int
            배치 크기
        """
        self.embedding = embedding
        self.X_train = X_train
        self.y_train = y_train
        self.X_test = X_test
        self.y_test = y_test
        self.max_len = max_len
        self.batch_size = batch_size
        self.pad_idx = pad_idx
        self.unk_idx = unk_idx

        self.dataloaders = {}
        self._create_all_dataloaders()

    def _create_all_dataloaders(self):
        """모든 임베딩에 대한 DataLoader 생성"""
        print("\n" + "="*70)
        print("DataLoader 생성 중...")
        print("="*70)

        # Word2Vec
        if 'word2vec' in self.embedding.models:
            self.dataloaders['word2vec'] = self._create_dataloader('word2vec')

        # FastText
        if 'fasttext' in self.embedding.models:
            self.dataloaders['fasttext'] = self._create_dataloader('fasttext')

        # GloVe
        if 'glove' in self.embedding.models:
            self.dataloaders['glove'] = self._create_dataloader('glove')

        print("\n✓ 모든 DataLoader 생성 완료!\n")

    def _create_dataloader(self, emb_name):
        """특정 임베딩에 대한 DataLoader 생성"""
        # word2idx 추출
        if emb_name in ['word2vec', 'fasttext']:
            word2idx = self.embedding.models[emb_name].wv.key_to_index
        else:  # glove
            word2idx = self.embedding.models[emb_name]['word_to_idx']

        # Dataset 생성 (✅ add_offset=True 추가)
        train_dataset = TextDataset(
            self.X_train, self.y_train, word2idx,
            self.max_len, self.pad_idx, self.unk_idx,
            add_offset=True  # ✅ 추가!
        )
        test_dataset = TextDataset(
            self.X_test, self.y_test, word2idx,
            self.max_len, self.pad_idx, self.unk_idx,
            add_offset=True  # ✅ 추가!
        )

        # DataLoader 생성
        train_loader = DataLoader(train_dataset, batch_size=self.batch_size, shuffle=True)
        test_loader = DataLoader(test_dataset, batch_size=self.batch_size, shuffle=False)

        print(f"  ✓ {emb_name.upper():10s} (vocab: {len(word2idx):5d} 단어)")

        return train_loader, test_loader

    def get_dataloader(self, emb_name):
        """특정 임베딩의 DataLoader 가져오기"""
        if emb_name not in self.dataloaders:
            raise ValueError(f"{emb_name}에 대한 DataLoader가 없습니다.")
        return self.dataloaders[emb_name]

    def get_all_dataloaders(self):
        """모든 DataLoader 반환"""
        return self.dataloaders

    def get_vocab_size(self, emb_name):
        """특정 임베딩의 vocabulary 크기"""
        if emb_name in ['word2vec', 'fasttext']:
            return len(self.embedding.models[emb_name].wv.key_to_index)
        else:  # glove
            return len(self.embedding.models[emb_name]['word_to_idx'])

    def show_info(self):
        """DataLoader 정보 출력"""
        print("\n" + "="*70)
        print("DataLoader 정보")
        print("="*70)

        for emb_name in self.dataloaders.keys():
            train_loader, test_loader = self.dataloaders[emb_name]
            vocab_size = self.get_vocab_size(emb_name)

            print(f"\n[{emb_name.upper()}]")
            print(f"  Vocabulary 크기: {vocab_size}")
            print(f"  Train batches: {len(train_loader)}")
            print(f"  Test batches:  {len(test_loader)}")
            print(f"  Batch size:    {self.batch_size}")
            print(f"  Max length:    {self.max_len}")
````

## 모델 구현

### LSTM(미사용)

단방향을 사용할 것이다. `bidirectional=False`로 설정해야하고 `bidirectional=True`처럼 양방향을 사용한다고 하면
```
# 양방향 (성능 향상 가능)
self.lstm = nn.LSTM(..., bidirectional=True)
self.fc = nn.Linear(hidden_dim * 2, output_dim)  # 차원 수정

# forward에서
hidden = torch.cat((hidden[-2], hidden[-1]), dim=1)  # 양방향 concat
output = self.fc(hidden)
```
를 수정해야한다.

````python
class LSTMModel(nn.Module):
    def __init__(self, embedding_matrix, hidden_dim, output_dim,
                 n_layers=2, bidirectional=False, dropout=0.3,
                 freeze_embeddings=False, pooling='last'):

        super(LSTMModel, self).__init__()

        self.hidden_dim = hidden_dim
        self.n_layers = n_layers
        self.bidirectional = bidirectional
        self.pooling = pooling

        # 임베딩 레이어 (사전 학습된 임베딩 직접 로드)
        self.embedding = nn.Embedding.from_pretrained(
            torch.tensor(embedding_matrix, dtype=torch.float),
            freeze=freeze_embeddings
        )

        embedding_dim = embedding_matrix.shape[1]

        # LSTM (bidirectional 파라미터 추가!)
        self.lstm = nn.LSTM(
            embedding_dim,
            hidden_dim,
            num_layers=n_layers,
            dropout=dropout if n_layers > 1 else 0,  # n_layers=1이면 dropout 무시
            batch_first=True,
            bidirectional=bidirectional  # ✅ 추가!
        )

        # bidirectional이면 hidden_dim이 2배
        lstm_output_dim = hidden_dim * 2 if bidirectional else hidden_dim

        # 드롭아웃
        self.dropout = nn.Dropout(dropout)

        # Fully Connected Layers
        self.fc1 = nn.Linear(lstm_output_dim, 128)  # ✅ lstm_output_dim 사용
        self.fc2 = nn.Linear(128, 64)
        self.fc3 = nn.Linear(64, output_dim)

    def forward(self, x):
        # 임베딩
        embedded = self.embedding(x)  # (batch, seq_len, embedding_dim)

        # LSTM
        lstm_out, (hidden, cell) = self.lstm(embedded)
        # lstm_out: (batch, seq_len, hidden_dim * num_directions)
        # hidden: (num_layers * num_directions, batch, hidden_dim)

        # Pooling 방식 선택
        if self.pooling == 'last':
            # 마지막 hidden state 사용
            if self.bidirectional:
                # forward와 backward의 마지막 hidden state concat
                # hidden[-2]: forward의 마지막 레이어
                # hidden[-1]: backward의 마지막 레이어
                output = torch.cat((hidden[-2], hidden[-1]), dim=1)
            else:
                output = hidden[-1]

        elif self.pooling == 'mean':
            # LSTM 출력의 평균
            output = torch.mean(lstm_out, dim=1)

        elif self.pooling == 'max':
            # LSTM 출력의 최댓값
            output = torch.max(lstm_out, dim=1)[0]

        else:
            raise ValueError(f"Unknown pooling method: {self.pooling}")

        # Fully Connected Layers
        output = self.dropout(output)
        output = F.relu(self.fc1(output))
        output = self.dropout(output)
        output = F.relu(self.fc2(output))
        output = self.dropout(output)
        output = self.fc3(output)

        return output
````

### 양방향 LSTM

````python
class BiLSTMModel(nn.Module):
    def __init__(self, embedding_matrix, hidden_dim, output_dim,
                 n_layers=2, dropout=0.3, freeze_embeddings=False, pooling='last'):
        super(BiLSTMModel, self).__init__()

        self.hidden_dim = hidden_dim
        self.n_layers = n_layers
        self.pooling = pooling

        # 임베딩 레이어
        self.embedding = nn.Embedding.from_pretrained(
            torch.tensor(embedding_matrix, dtype=torch.float),
            freeze=freeze_embeddings
        )

        embedding_dim = embedding_matrix.shape[1]

        # 양방향 LSTM
        self.lstm = nn.LSTM(
            embedding_dim,
            hidden_dim,
            num_layers=n_layers,
            dropout=dropout if n_layers > 1 else 0,
            batch_first=True,
            bidirectional=True  # ✅ 양방향!
        )

        # 양방향이므로 출력 차원 2배
        lstm_output_dim = hidden_dim * 2

        # 드롭아웃
        self.dropout = nn.Dropout(dropout)

        # Fully Connected Layers
        self.fc1 = nn.Linear(lstm_output_dim, 128)  # ✅ 2배 차원
        self.fc2 = nn.Linear(128, 64)
        self.fc3 = nn.Linear(64, output_dim)

    def forward(self, x):
        # 임베딩
        embedded = self.embedding(x)

        # 양방향 LSTM
        lstm_out, (hidden, cell) = self.lstm(embedded)
        # lstm_out: (batch, seq_len, hidden_dim * 2)
        # hidden: (n_layers * 2, batch, hidden_dim)

        # Pooling
        if self.pooling == 'last':
            # Forward와 Backward의 마지막 hidden state concat
            # hidden[-2]: forward의 마지막 레이어
            # hidden[-1]: backward의 마지막 레이어
            output = torch.cat((hidden[-2], hidden[-1]), dim=1)

        elif self.pooling == 'mean':
            # LSTM 출력의 평균
            output = torch.mean(lstm_out, dim=1)

        elif self.pooling == 'max':
            # LSTM 출력의 최댓값
            output = torch.max(lstm_out, dim=1)[0]
        else:
            raise ValueError(f"Unknown pooling: {self.pooling}")

        # FC Layers
        output = self.dropout(output)
        output = F.relu(self.fc1(output))
        output = self.dropout(output)
        output = F.relu(self.fc2(output))
        output = self.dropout(output)
        output = self.fc3(output)

        return output
````

### GRU

````python
class GRUModel(nn.Module):

    def __init__(self, embedding_matrix, hidden_dim, output_dim,
                 n_layers=2, dropout=0.3, freeze_embeddings=False, pooling='last'):
        super(GRUModel, self).__init__()

        self.hidden_dim = hidden_dim
        self.n_layers = n_layers
        self.pooling = pooling

        # 임베딩 레이어
        self.embedding = nn.Embedding.from_pretrained(
            torch.tensor(embedding_matrix, dtype=torch.float),
            freeze=freeze_embeddings,
            padding_idx=0
        )

        embedding_dim = embedding_matrix.shape[1]

        # GRU 레이어
        self.gru = nn.GRU(
            embedding_dim,
            hidden_dim,
            num_layers=n_layers,
            dropout=dropout if n_layers > 1 else 0,
            batch_first=True
        )

        # 드롭아웃
        self.dropout = nn.Dropout(dropout)

        # Fully Connected Layers
        self.fc1 = nn.Linear(hidden_dim, 128)
        self.fc2 = nn.Linear(128, 64)
        self.fc3 = nn.Linear(64, output_dim)

    def forward(self, x):

        # 임베딩
        embedded = self.embedding(x)  # (batch, seq_len, embedding_dim)

        # GRU
        gru_out, hidden = self.gru(embedded)
        # gru_out: (batch, seq_len, hidden_dim)
        # hidden: (n_layers, batch, hidden_dim)

        # Pooling 방식 적용
        if self.pooling == 'last':
            # 마지막 hidden state 사용
            output = hidden[-1]

        elif self.pooling == 'mean':
            # 시퀀스 전체의 평균
            output = torch.mean(gru_out, dim=1)

        elif self.pooling == 'max':
            # 시퀀스 전체의 최댓값
            output = torch.max(gru_out, dim=1)[0]

        else:
            raise ValueError(f"Unknown pooling method: {self.pooling}")

        # Fully Connected Layers
        output = self.dropout(output)
        output = F.relu(self.fc1(output))
        output = self.dropout(output)
        output = F.relu(self.fc2(output))
        output = self.dropout(output)
        output = self.fc3(output)

        return output

    def get_model_info(self):
        """모델 정보 출력"""
        total_params = sum(p.numel() for p in self.parameters())
        trainable_params = sum(p.numel() for p in self.parameters() if p.requires_grad)

        info = {
            'Model Type': 'GRU',
            'Hidden Dim': self.hidden_dim,
            'Num Layers': self.n_layers,
            'Pooling': self.pooling,
            'Total Parameters': f"{total_params:,}",
            'Trainable Parameters': f"{trainable_params:,}"
        }

        return info
````

## 모델 학습 및 평가

과제의 목표가 3개의 임베딩 모델에서 RNN기반 모델들로 최고 성능의 조합을 찾는 것이다. 그래서 다음과 같은 목표를 정했다.

1. 3가지 임베딩과 3가지 모델의 최고성능 조합을 찾아 내는 것이 목적
2. 모델별로 각 임베딩 별 결과를 시각화(그래프 텍스트는 영어로 출력)하는 클래스 또는 함수 작성
3. 생성된 9개의 결과중에 어떤 조합이 가장 성능이 뛰어난지 비교(결과에는 임베딩 모델, 딥러닝 모델의 파라미터 등이 텍스트로 출력되도록 코드 작성)
4. 평가지표는 Accuracy, Precision, Recall, F1-score을 사용할 것.  그중 주요 지표는 F1-score로 설정.(균형 있게 좋은 성능을 출력하는 모델 찾기 위해.)

````python
class EmbeddingModelEvaluator:
    """3가지 임베딩 × 3가지 모델 = 9개 조합 평가 시스템"""

    def __init__(self, embedding_obj, dl_manager, class_names, device='cuda'):
        """
        Parameters:
        -----------
        embedding_obj : MultiEmbedding
            학습된 임베딩 객체
        dl_manager : EmbeddingDataLoaderManager
            DataLoader 관리 객체
        class_names : list
            클래스 이름 리스트 (20개 뉴스 카테고리)
        device : str
            'cuda' or 'cpu'
        """
        self.embedding_obj = embedding_obj
        self.dl_manager = dl_manager
        self.class_names = class_names
        self.device = torch.device(device if torch.cuda.is_available() else 'cpu')

        self.embedding_names = ['word2vec', 'fasttext', 'glove']
        self.model_classes = {
            'LSTM': LSTMModel,
            'BiLSTM': BiLSTMModel,
            'GRU': GRUModel
        }

        self.results = {}
        self.embedding_matrices = {}

    def create_embedding_matrix(self, emb_name):
        """임베딩 행렬 생성"""
        if emb_name in self.embedding_matrices:
            return self.embedding_matrices[emb_name]

        if emb_name in ['word2vec', 'fasttext']:
            word2idx = self.embedding_obj.models[emb_name].wv.key_to_index
            vocab_size = len(word2idx) + 2
            embedding_dim = self.embedding_obj.vector_size

            embedding_matrix = np.zeros((vocab_size, embedding_dim))
            for word, idx in word2idx.items():
                embedding_matrix[idx + 2] = self.embedding_obj.models[emb_name].wv[word]

        else:  # glove
            word2idx = self.embedding_obj.models[emb_name]['word_to_idx']
            word_to_vec = self.embedding_obj.models[emb_name]['word_to_vec']
            vocab_size = len(word2idx) + 2
            embedding_dim = self.embedding_obj.vector_size

            embedding_matrix = np.zeros((vocab_size, embedding_dim))
            for word, idx in word2idx.items():
                embedding_matrix[idx + 2] = word_to_vec[word]

        self.embedding_matrices[emb_name] = embedding_matrix
        return embedding_matrix

    def train_epoch(self, model, train_loader, criterion, optimizer):
        """1 에포크 학습"""
        model.train()
        total_loss = 0

        for inputs, labels in train_loader:
            inputs, labels = inputs.to(self.device), labels.to(self.device)

            optimizer.zero_grad()
            outputs = model(inputs)
            loss = criterion(outputs, labels)
            loss.backward()
            optimizer.step()

            total_loss += loss.item()

        return total_loss / len(train_loader)

    def evaluate(self, model, test_loader, return_predictions=False):
        """모델 평가 - 모든 지표 계산"""
        model.eval()
        all_preds = []
        all_labels = []

        with torch.no_grad():
            for inputs, labels in test_loader:
                inputs, labels = inputs.to(self.device), labels.to(self.device)
                outputs = model(inputs)
                _, predicted = torch.max(outputs, 1)

                all_preds.extend(predicted.cpu().numpy())
                all_labels.extend(labels.cpu().numpy())

        # 전체 평가 지표
        accuracy = accuracy_score(all_labels, all_preds)
        precision, recall, f1, _ = precision_recall_fscore_support(
            all_labels, all_preds, average='weighted', zero_division=0
        )

        # 클래스별 평가 지표
        class_precision, class_recall, class_f1, class_support = precision_recall_fscore_support(
            all_labels, all_preds, average=None, zero_division=0
        )

        # Confusion Matrix
        conf_matrix = confusion_matrix(all_labels, all_preds)

        metrics = {
            'accuracy': accuracy * 100,
            'precision': precision * 100,
            'recall': recall * 100,
            'f1': f1 * 100,
            'class_metrics': {
                'precision': class_precision * 100,
                'recall': class_recall * 100,
                'f1': class_f1 * 100,
                'support': class_support
            },
            'confusion_matrix': conf_matrix
        }

        if return_predictions:
            metrics['predictions'] = all_preds
            metrics['labels'] = all_labels

        return metrics

    def train_single_model(self, emb_name, model_name, epochs=10, lr=0.001):
        """단일 조합 학습"""
        print(f"\n{'='*70}")
        print(f"Training: {emb_name.upper()} + {model_name}")
        print(f"{'='*70}")

        embedding_matrix = self.create_embedding_matrix(emb_name)
        train_loader, test_loader = self.dl_manager.get_dataloader(emb_name)

        model_class = self.model_classes[model_name]

        if model_name == 'GRU':
            model = model_class(
                embedding_matrix=embedding_matrix,
                hidden_dim=256,
                output_dim=20,
                n_layers=2,
                dropout=0.3,
                freeze_embeddings=True,
                pooling='last'
            )
        elif model_name == 'BiLSTM':
            model = model_class(
                embedding_matrix=embedding_matrix,
                hidden_dim=256,
                output_dim=20,
                n_layers=2,
                dropout=0.3,
                freeze_embeddings=True,
                pooling='last'
            )
        else:  # LSTM
            model = model_class(
                embedding_matrix=embedding_matrix,
                hidden_dim=256,         # ✅ 256
                output_dim=20,
                n_layers=2,             # ✅ 2
                bidirectional=False,
                dropout=0.3,            # ✅ 0.3
                freeze_embeddings=True,
                pooling='last'
            )

        model = model.to(self.device)
        # 손실함수 가중치 제거
        criterion = nn.CrossEntropyLoss()
        optimizer = optim.Adam(model.parameters(), lr=lr)

        best_f1 = 0
        best_metrics = None

        for epoch in range(epochs):
            train_loss = self.train_epoch(model, train_loader, criterion, optimizer)
            metrics = self.evaluate(model, test_loader, return_predictions=True)

            if metrics['f1'] > best_f1:
                best_f1 = metrics['f1']
                best_metrics = metrics.copy()

            if (epoch + 1) % 2 == 0:
                print(f"Epoch {epoch+1}/{epochs} - Loss: {train_loss:.4f}, "
                      f"F1: {metrics['f1']:.2f}%, Acc: {metrics['accuracy']:.2f}%")

        total_params = sum(p.numel() for p in model.parameters())

        key = f"{emb_name}_{model_name}"
        self.results[key] = {
            'embedding': emb_name,
            'model': model_name,
            'metrics': best_metrics,
            'params': total_params
        }

        print(f"✓ Best F1: {best_f1:.2f}%")
        return best_metrics

    def train_all_combinations(self, epochs=10, lr=0.001):
        """모든 9개 조합 학습"""
        print("\n" + "="*70)
        print("Training All 9 Combinations")
        print("="*70)

        for emb_name in self.embedding_names:
            for model_name in self.model_classes.keys():
                self.train_single_model(emb_name, model_name, epochs, lr)

        print("\n✓ All combinations trained!")

    def get_results_dataframe(self):
        """결과를 DataFrame으로 반환"""
        data = []
        for key, result in self.results.items():
            row = {
                'Embedding': result['embedding'].upper(),
                'Model': result['model'],
                'Accuracy': result['metrics']['accuracy'],
                'Precision': result['metrics']['precision'],
                'Recall': result['metrics']['recall'],
                'F1-Score': result['metrics']['f1'],
                'Parameters': result['params']
            }
            data.append(row)

        df = pd.DataFrame(data)
        df = df.sort_values('F1-Score', ascending=False)
        return df

    def plot_results(self):
        """결과 시각화"""
        df = self.get_results_dataframe()

        fig, axes = plt.subplots(2, 2, figsize=(16, 12))

        # 1. F1-Score 비교
        ax = axes[0, 0]
        df_sorted = df.sort_values('F1-Score', ascending=True)
        colors = plt.cm.viridis(np.linspace(0, 1, len(df_sorted)))
        ax.barh(range(len(df_sorted)), df_sorted['F1-Score'], color=colors)
        ax.set_yticks(range(len(df_sorted)))
        ax.set_yticklabels([f"{row['Embedding']}+{row['Model']}"
                            for _, row in df_sorted.iterrows()])
        ax.set_xlabel('F1-Score (%)')
        ax.set_title('F1-Score Comparison (All Combinations)')
        ax.grid(axis='x', alpha=0.3)

        # 2. 임베딩별 성능 비교
        ax = axes[0, 1]
        for emb in ['WORD2VEC', 'FASTTEXT', 'GLOVE']:
            emb_data = df[df['Embedding'] == emb]
            ax.plot(emb_data['Model'], emb_data['F1-Score'],
                   marker='o', label=emb, linewidth=2)
        ax.set_xlabel('Model')
        ax.set_ylabel('F1-Score (%)')
        ax.set_title('F1-Score by Embedding Type')
        ax.legend()
        ax.grid(True, alpha=0.3)

        # 3. 모델별 성능 비교
        ax = axes[1, 0]
        for model in ['BiLSTM', 'GRU']:
            model_data = df[df['Model'] == model]
            ax.plot(model_data['Embedding'], model_data['F1-Score'],
                   marker='s', label=model, linewidth=2)
        ax.set_xlabel('Embedding')
        ax.set_ylabel('F1-Score (%)')
        ax.set_title('F1-Score by Model Type')
        ax.legend()
        ax.grid(True, alpha=0.3)

        # 4. Heatmap
        ax = axes[1, 1]
        pivot = df.pivot(index='Model', columns='Embedding', values='F1-Score')
        sns.heatmap(pivot, annot=True, fmt='.2f', cmap='YlOrRd', ax=ax,
                   cbar_kws={'label': 'F1-Score (%)'})
        ax.set_title('F1-Score Heatmap')

        plt.tight_layout()
        plt.savefig('./embedding_model_comparison.png', dpi=300, bbox_inches='tight')
        plt.show()

    def plot_metrics_comparison(self):
        """모든 평가 지표 비교"""
        df = self.get_results_dataframe()

        fig, axes = plt.subplots(2, 2, figsize=(16, 12))
        metrics = ['Accuracy', 'Precision', 'Recall', 'F1-Score']

        for idx, metric in enumerate(metrics):
            ax = axes[idx // 2, idx % 2]
            df_sorted = df.sort_values(metric, ascending=True)
            colors = plt.cm.RdYlGn(np.linspace(0.3, 0.9, len(df_sorted)))
            ax.barh(range(len(df_sorted)), df_sorted[metric], color=colors)
            ax.set_yticks(range(len(df_sorted)))
            ax.set_yticklabels([f"{row['Embedding']}+{row['Model']}"
                               for _, row in df_sorted.iterrows()], fontsize=9)
            ax.set_xlabel(f'{metric} (%)')
            ax.set_title(f'{metric} Comparison')
            ax.grid(axis='x', alpha=0.3)

        plt.tight_layout()
        plt.savefig('./all_metrics_comparison.png', dpi=300, bbox_inches='tight')
        plt.show()

    def plot_confusion_matrices(self, top_n=3):
        """상위 N개 조합의 Confusion Matrix 시각화"""
        df = self.get_results_dataframe()
        top_combinations = df.head(top_n)

        fig, axes = plt.subplots(1, top_n, figsize=(8*top_n, 6))
        if top_n == 1:
            axes = [axes]

        for idx, (_, row) in enumerate(top_combinations.iterrows()):
            key = f"{row['Embedding'].lower()}_{row['Model']}"
            conf_matrix = self.results[key]['metrics']['confusion_matrix']

            ax = axes[idx]
            sns.heatmap(conf_matrix, annot=False, fmt='d', cmap='Blues', ax=ax,
                       xticklabels=self.class_names, yticklabels=self.class_names)
            ax.set_title(f"{row['Embedding']}+{row['Model']}\nF1: {row['F1-Score']:.2f}%")
            ax.set_xlabel('Predicted')
            ax.set_ylabel('Actual')
            plt.setp(ax.get_xticklabels(), rotation=45, ha='right', fontsize=8)
            plt.setp(ax.get_yticklabels(), rotation=0, fontsize=8)

        plt.tight_layout()
        plt.savefig('./confusion_matrices_top.png', dpi=300, bbox_inches='tight')
        plt.show()

    def plot_class_performance(self, combination_key=None):
        """특정 조합의 클래스별 성능 시각화"""
        if combination_key is None:
            # 최고 성능 조합 선택
            df = self.get_results_dataframe()
            best = df.iloc[0]
            combination_key = f"{best['Embedding'].lower()}_{best['Model']}"

        metrics = self.results[combination_key]['metrics']
        class_metrics = metrics['class_metrics']

        fig, axes = plt.subplots(2, 2, figsize=(16, 12))

        # 1. 클래스별 F1-Score
        ax = axes[0, 0]
        f1_scores = class_metrics['f1']
        colors = plt.cm.RdYlGn(np.linspace(0.3, 0.9, len(f1_scores)))
        sorted_indices = np.argsort(f1_scores)
        ax.barh(range(len(f1_scores)), f1_scores[sorted_indices], color=colors[sorted_indices])
        ax.set_yticks(range(len(f1_scores)))
        ax.set_yticklabels([self.class_names[i] for i in sorted_indices], fontsize=9)
        ax.set_xlabel('F1-Score (%)')
        ax.set_title('F1-Score by Class')
        ax.grid(axis='x', alpha=0.3)

        # 2. 클래스별 Precision vs Recall
        ax = axes[0, 1]
        precision = class_metrics['precision']
        recall = class_metrics['recall']
        ax.scatter(precision, recall, s=100, alpha=0.6)
        for i, name in enumerate(self.class_names):
            ax.annotate(name, (precision[i], recall[i]), fontsize=7, alpha=0.7)
        ax.set_xlabel('Precision (%)')
        ax.set_ylabel('Recall (%)')
        ax.set_title('Precision vs Recall by Class')
        ax.grid(True, alpha=0.3)
        ax.plot([0, 100], [0, 100], 'k--', alpha=0.3)

        # 3. 클래스별 Support (샘플 수)
        ax = axes[1, 0]
        support = class_metrics['support']
        ax.bar(range(len(support)), support, color='skyblue')
        ax.set_xticks(range(len(support)))
        ax.set_xticklabels(self.class_names, rotation=45, ha='right', fontsize=9)
        ax.set_ylabel('Number of Samples')
        ax.set_title('Test Samples by Class')
        ax.grid(axis='y', alpha=0.3)

        # 4. 클래스별 종합 메트릭 비교
        ax = axes[1, 1]
        x = np.arange(len(self.class_names))
        width = 0.25
        ax.bar(x - width, class_metrics['precision'], width, label='Precision', alpha=0.8)
        ax.bar(x, class_metrics['recall'], width, label='Recall', alpha=0.8)
        ax.bar(x + width, class_metrics['f1'], width, label='F1-Score', alpha=0.8)
        ax.set_xticks(x)
        ax.set_xticklabels(self.class_names, rotation=45, ha='right', fontsize=9)
        ax.set_ylabel('Score (%)')
        ax.set_title('Metrics Comparison by Class')
        ax.legend()
        ax.grid(axis='y', alpha=0.3)

        plt.suptitle(f"{self.results[combination_key]['embedding'].upper()} + "
                    f"{self.results[combination_key]['model']}", fontsize=14, y=1.00)
        plt.tight_layout()
        plt.savefig(f'./class_performance_{combination_key}.png', dpi=300, bbox_inches='tight')
        plt.show()

    def plot_all_class_performance(self):
        """모든 9개 조합의 클래스별 평균 성능"""
        all_f1_scores = []
        combinations = []

        for key, result in self.results.items():
            class_f1 = result['metrics']['class_metrics']['f1']
            all_f1_scores.append(class_f1)
            combinations.append(f"{result['embedding'].upper()}+{result['model']}")

        all_f1_scores = np.array(all_f1_scores)

        fig, axes = plt.subplots(1, 2, figsize=(18, 6))

        # 1. 히트맵 - 조합별 클래스별 F1-Score
        ax = axes[0]
        sns.heatmap(all_f1_scores, annot=False, cmap='RdYlGn', ax=ax,
                   xticklabels=self.class_names, yticklabels=combinations,
                   cbar_kws={'label': 'F1-Score (%)'}, vmin=0, vmax=100)
        ax.set_title('Class-wise F1-Score Heatmap (All Combinations)')
        ax.set_xlabel('Class')
        ax.set_ylabel('Combination')
        plt.setp(ax.get_xticklabels(), rotation=45, ha='right', fontsize=9)

        # 2. 클래스별 평균 성능 (9개 조합 평균)
        ax = axes[1]
        mean_f1 = all_f1_scores.mean(axis=0)
        std_f1 = all_f1_scores.std(axis=0)
        sorted_indices = np.argsort(mean_f1)

        colors = plt.cm.RdYlGn(np.linspace(0.3, 0.9, len(mean_f1)))
        ax.barh(range(len(mean_f1)), mean_f1[sorted_indices],
               xerr=std_f1[sorted_indices], color=colors[sorted_indices], alpha=0.8)
        ax.set_yticks(range(len(mean_f1)))
        ax.set_yticklabels([self.class_names[i] for i in sorted_indices], fontsize=9)
        ax.set_xlabel('Average F1-Score (%)')
        ax.set_title('Average Class Performance (Across All Combinations)')
        ax.grid(axis='x', alpha=0.3)

        plt.tight_layout()
        plt.savefig('./all_combinations_class_performance.png', dpi=300, bbox_inches='tight')
        plt.show()

    def print_class_report(self, combination_key=None):
        """특정 조합의 클래스별 상세 리포트"""
        if combination_key is None:
            df = self.get_results_dataframe()
            best = df.iloc[0]
            combination_key = f"{best['Embedding'].lower()}_{best['Model']}"

        result = self.results[combination_key]
        class_metrics = result['metrics']['class_metrics']

        print("\n" + "="*100)
        print(f"CLASS-WISE PERFORMANCE: {result['embedding'].upper()} + {result['model']}")
        print("="*100)
        print(f"{'Class':<25}{'Precision':<12}{'Recall':<12}{'F1-Score':<12}{'Support':<12}")
        print("-"*100)

        for i, name in enumerate(self.class_names):
            print(f"{name:<25}{class_metrics['precision'][i]:<12.2f}"
                  f"{class_metrics['recall'][i]:<12.2f}"
                  f"{class_metrics['f1'][i]:<12.2f}"
                  f"{int(class_metrics['support'][i]):<12}")

        print("-"*100)
        print(f"{'OVERALL':<25}{result['metrics']['precision']:<12.2f}"
              f"{result['metrics']['recall']:<12.2f}"
              f"{result['metrics']['f1']:<12.2f}"
              f"{int(class_metrics['support'].sum()):<12}")
        print("="*100)

    def print_best_combination(self):
        """최고 성능 조합 출력"""
        df = self.get_results_dataframe()
        best = df.iloc[0]

        print("\n" + "="*70)
        print("BEST COMBINATION (Based on F1-Score)")
        print("="*70)
        print(f"\n🏆 Best Embedding: {best['Embedding']}")
        print(f"🏆 Best Model:     {best['Model']}")
        print(f"\n📊 Performance Metrics:")
        print(f"   • Accuracy:  {best['Accuracy']:.2f}%")
        print(f"   • Precision: {best['Precision']:.2f}%")
        print(f"   • Recall:    {best['Recall']:.2f}%")
        print(f"   • F1-Score:  {best['F1-Score']:.2f}%")
        print(f"\n⚙️  Model Parameters: {best['Parameters']:,}")
        print("="*70)

    def print_full_results_table(self):
        """전체 결과 테이블 출력"""
        df = self.get_results_dataframe()

        print("\n" + "="*100)
        print("FULL RESULTS TABLE (Sorted by F1-Score)")
        print("="*100)
        print(f"{'Rank':<6}{'Embedding':<12}{'Model':<10}{'Acc(%)':<10}{'Prec(%)':<10}"
              f"{'Rec(%)':<10}{'F1(%)':<10}{'Params':<15}")
        print("-"*100)

        for idx, (_, row) in enumerate(df.iterrows(), 1):
            print(f"{idx:<6}{row['Embedding']:<12}{row['Model']:<10}"
                  f"{row['Accuracy']:<10.2f}{row['Precision']:<10.2f}"
                  f"{row['Recall']:<10.2f}{row['F1-Score']:<10.2f}"
                  f"{row['Parameters']:<15,}")

        print("="*100)

    def save_results(self, filepath='./results.csv'):
        """결과를 CSV로 저장"""
        df = self.get_results_dataframe()
        df.to_csv(filepath, index=False)
        print(f"\n✓ Results saved to {filepath}")
````

- "get"의 Word2Vec 인덱스가 5라면
- 임베딩 행렬: 7번 위치에 "get"의 벡터 저장

### 1차 수정 결과(인덱스 매핑문제 해결)

기존에 임배딩행렬의 매핑과 데이터로더의 매핑이 불일치 한경우가 있었다.

**🎯 왜 이 문제가 발생했는가?**

1. 임베딩 행렬: PAD(0), UNK(1)을 위해 처음 2개 위치 예약
2. 인코딩: Word2Vec의 원본 인덱스를 그대로 사용
3. 결과: 2칸 차이로 완전히 다른 단어 참조!

```python
class TextDataset(Dataset):
    def __init__(self, texts, labels, word2idx, max_len,
                 pad_idx=0, unk_idx=1, lowercase=True, add_offset=True):
        self.texts = texts
        self.labels = labels
        self.lowercase = lowercase
        self.max_len = max_len
        self.pad_idx = pad_idx
        self.unk_idx = unk_idx
        self.add_offset = add_offset
        
        # ✅ word2idx를 미리 +2 오프셋으로 변환
        if add_offset:
            self.word2idx = {word: idx + 2 for word, idx in word2idx.items()}
        else:
            self.word2idx = word2idx

    def __len__(self):
        return len(self.texts)

    def __getitem__(self, idx):
        text = self.texts[idx]
        
        if self.lowercase:
            text = text.lower()

        tokens = word_tokenize(text)
        encoded = [self.word2idx.get(word, self.unk_idx) for word in tokens]

        if len(encoded) < self.max_len:
            encoded += [self.pad_idx] * (self.max_len - len(encoded))
        else:
            encoded = encoded[:self.max_len]

        encoded_tensor = torch.tensor(encoded, dtype=torch.long)
        label_tensor = torch.tensor(self.labels[idx], dtype=torch.long)

        return encoded_tensor, label_tensor
```

### 3차 수정: 클래스 가중치 제거, GLoVe 사전크기 조정, glove_max_iter 조정, 옵티마이저 AdamW사용, LSTM 원점 회귀

1. 클래스 가중치 제거: 데이터 불균형이 심하지 않아 없는 것이 더 효과적
2. glove_max_iter 조정, GLoVe 사전크기 조정: 전반적인 성능은 향상되었다. 하지만 GRU에서 성능이 하락이 되었고, 단어 사전의 크기가 커짐에 따라 많은 단어를 학습 할 수 있지만, 희귀 단어의 출현 빈도가 낮았고, 임베딩 품질이 저하되어 노이즈에 모델이 민감하다고 예상된다. 약간은 크기를 줄여서 학습하는 것이 좋을 것 같다.
3. 옵티마이저 AdamW사용: 데이터가 크지 않아 오히려 임베딩 가중치도 규제하는 것 같다. 이로서 종보 손실이 발생하여 성능이 저하 된 것 같다. Adam을 사용하는 것이 좋아보인다.
4. LSTM 원점 회귀: 성능은 향상되었으나, 여전히 낮은 성능을 보여준다. 따라서 제외하는 것이 맞는 것 같다.
5. 부족한 학습 횟수: 현재 20회의 에폭으로 학습중인데 이것으로 인해 충분히 학습하지 못해 수렴이 좀처럼 이뤄지지 않고 있다. 30까지 늘려 경과를 보자.
