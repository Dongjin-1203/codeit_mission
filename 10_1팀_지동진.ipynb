````python
# 5단계: 불용어가 아닌 단어만 필터링
tokens = [word for word in tokens if word not in stop_words]
print("4단계:", tokens)
````

**전체 전처리 코드**

````python
def clean_text(text):
    text = text.lower() # 소문자로 변경
    text = re.sub(r'[^a-zA-Z\s]', '', text) # 공백과 알파벳을 제외한 모든 것을 제거
    stop_words = set(stopwords.words('english'))
    tokens = text.split()
    tokens = [word for word in tokens if word not in stop_words]
    return ' '.join(tokens)
````

````python
# 데이터에 적용
X_train_processed = [clean_text(text) for text in X_train]
X_test_processed = [clean_text(text) for text in X_test]
````

### 2. 텍스트 임베딩

````python
# word2vec 모델 선언
word2vec = Word2Vec(
    sentences=train_sentences,
    vector_size=128,    # 각 단어를 표현하는 벡터의 차원 수
    window=5,   # 중심 단어 기준으로 좌우 몇 개의 단어를 문맥으로 볼지 결정
    min_count=2,    # 최소 등장 횟수
    sg=0,   # 0은 CBOW, 1은 Skip-gram
    epochs=10,  # 전체 데이터셋을 몇 번 반복해서 학습할지
    max_final_vocab=12000   # 최종 어휘 사전의 최대 크기
)
# 임베딩 매트리스 초기화
word2vec_matrix = np.zeros((len(word2vec.wv) + 2, 128))  # +2 (PAD, UNK)
# 단어-인덱스 매핑 딕셔너리 생성
word2idx_word2vec = {word: idx + 2 for idx, word in enumerate(word2vec.wv.index_to_key)}

# 임베딩 매트릭스에 단어 벡터 채우기
for word, idx in word2idx_word2vec.items():
    word2vec_matrix[idx] = word2vec.wv[word]
````

## 데이터셋 및 로더 생성

#### pytorch 데이터셋 생성

- texts (list): 텍스트 리스트     
- labels (list): 레이블 리스트        
- word2idx (dict): 단어를 인덱스로 매핑하는 딕셔너리      
- max_len (int): 최대 시퀀스 길이     
- pad_idx (int): 패딩 토큰 인덱스 (기본값: 0)     
- unk_idx (int): Unknown 토큰 인덱스 (기본값: 1)          
- lowercase (bool): 소문자 변환 여부 (기본값: True)

````python
class TextDataset(Dataset):
    def __init__(self, texts, labels, word2idx, max_len,
                 pad_idx=0, unk_idx=1, lowercase=True ):
        self.texts = texts
        self.labels = labels
        self.lowercase = lowercase
        self.word2idx = word2idx
        self.max_len = max_len
        self.pad_idx = pad_idx
        self.unk_idx = unk_idx

    def __len__(self):
        return len(self.texts)

    def __getitem__(self, idx):
        # 텍스트 가져오기
        text = self.texts[idx]

        # 토큰화
        tokens = word_tokenize(text)

        # 인덱스로 변환 (OOV 단어는 unk_idx)
        encoded = [self.word2idx.get(word, self.unk_idx) for word in tokens]

        # 패딩 또는 트렁케이션
        if len(encoded) < self.max_len:
            # 패딩 추가
            encoded += [self.pad_idx] * (self.max_len - len(encoded))
        else:
            # 최대 길이로 자르기
            encoded = encoded[:self.max_len]

        # 텐서로 변환
        encoded_tensor = torch.tensor(encoded, dtype=torch.long)
        label_tensor = torch.tensor(self.labels[idx], dtype=torch.long)

        return encoded_tensor, label_tensor
````

### LSTM

단방향을 사용할 것이다. `bidirectional=False`로 설정해야하고 `bidirectional=True`처럼 양방향을 사용한다고 하면
```
# 양방향 (성능 향상 가능)
self.lstm = nn.LSTM(..., bidirectional=True)
self.fc = nn.Linear(hidden_dim * 2, output_dim)  # 차원 수정

# forward에서
hidden = torch.cat((hidden[-2], hidden[-1]), dim=1)  # 양방향 concat
output = self.fc(hidden)
```
를 수정해야한다.

````python
class LSTMModel(nn.Module):
    def __init__(self, embedding_matrix, hidden_dim, output_dim,
                 n_layers=2, bidirectional=False, dropout=0.3,
                 freeze_embeddings=False, pooling='last'):
        super(LSTMModel, self).__init__()

        # 임베딩 레이어 (사전 학습된 임베딩 직접 로드)
        self.embedding = nn.Embedding.from_pretrained(
            torch.tensor(embedding_matrix, dtype=torch.float),
            freeze=freeze_embeddings
        )

        embedding_dim = embedding_matrix.shape[1]

        # LSTM
        self.lstm = nn.LSTM(embedding_dim, hidden_dim, num_layers=n_layers,
                           dropout=dropout, batch_first=True)

        # 드롭아웃
        self.dropout = nn.Dropout(dropout)

        # Fully connected Layer (3층으로 수정)
        self.fc1 = nn.Linear(hidden_dim, 128)  # 256 → 128
        self.fc2 = nn.Linear(128, 64)          # 128 → 64
        self.fc3 = nn.Linear(64, output_dim)   # 64 → 20

    def forward(self, x):
        # 임베딩
        embedded = self.embedding(x)

        # LSTM
        lstm_out, (hidden, cell) = self.lstm(embedded)

        # 마지막 hidden state 사용
        output = self.dropout(hidden[-1])
        output = F.relu(self.fc1(output))
        output = self.dropout(output)
        output = F.relu(self.fc2(output))
        output = self.dropout(output)
        output = self.fc3(output)

        return output
````

## 모델 학습 및 평가

````python
def train(model, loader, criterion, optimizer):
    model.train()
    total_loss = 0
    for texts, labels in loader:
        texts, labels = texts.to(device), labels.to(device)
        optimizer.zero_grad()
        outputs = model(texts)  # 출력: (batch_size, 20)
        loss = criterion(outputs, labels)  # 레이블은 정수형 (0~19)
        loss.backward()
        optimizer.step()
        total_loss += loss.item()
    return total_loss / len(loader)
````

````python
def evaluate(model, loader):
    model.eval()
    all_predictions = []
    all_labels = []

    with torch.no_grad():
        for texts, labels in loader:
            texts, labels = texts.to(device), labels.to(device)
            outputs = model(texts)
            predictions = torch.argmax(outputs, dim=1)  # 가장 높은 확률의 인덱스

            all_predictions.extend(predictions.cpu().numpy())
            all_labels.extend(labels.cpu().numpy())

    # 1. 전체 지표 출력
    print("=" * 50)
    print("Classification Report")
    print("=" * 50)
    print(classification_report(all_labels, all_predictions,
                                target_names=test_news.target_names))

    # 2. 개별 지표
    print("\n개별 지표 (Weighted Average)")
    print(f"Accuracy:  {accuracy_score(all_labels, all_predictions):.4f}")
    print(f"Precision: {precision_score(all_labels, all_predictions, average='weighted'):.4f}")
    print(f"Recall:    {recall_score(all_labels, all_predictions, average='weighted'):.4f}")
    print(f"F1-Score:  {f1_score(all_labels, all_predictions, average='weighted'):.4f}")

    # 3. Macro/Micro 평균
    print(f"\nMacro F1:  {f1_score(all_labels, all_predictions, average='macro'):.4f}")
    print(f"Micro F1:  {f1_score(all_labels, all_predictions, average='micro'):.4f}")

    # 4. 혼동행렬 시각화
    cm = confusion_matrix(all_labels, all_predictions)
    plt.figure(figsize=(12, 10))
    sns.heatmap(cm, annot=False, fmt='d', cmap='Blues')
    plt.xlabel('Predicted')
    plt.ylabel('Actual')
    plt.title('Confusion Matrix')
    plt.show()

    return all_predictions, all_labels
````

````python
# 검증 손실 계산 함수 추가
def validate(model, loader, criterion):
    model.eval()
    total_loss = 0
    with torch.no_grad():
        for texts, labels in loader:
            texts, labels = texts.to(device), labels.to(device)
            outputs = model(texts)
            loss = criterion(outputs, labels)
            total_loss += loss.item()
    return total_loss / len(loader)
````

````python
model_word2vec = LSTMModel(word2vec_matrix, hidden_dim, output_dim,
                           n_layers=2, dropout=0.3, freeze_embeddings = False).to(device)
# 1. Label Smoothing 추가
loss_fn = nn.CrossEntropyLoss(label_smoothing=0.1)
# 2. AdamW (Weight Decay 포함)
optimizer = torch.optim.AdamW(
    model_word2vec.parameters(),
    lr=0.003,
    weight_decay=0.01  # L2 regularization
)
# Learning rate scheduler 추가
scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='min', patience=3, factor=0.5)

# 학습 기록을 저장할 리스트
train_losses = []
val_losses = []
learning_rates = []

# 학습 루프
for epoch in range(50):
    train_loss = train(model_word2vec, train_loader_word2vec, loss_fn, optimizer)
    val_loss = validate(model_word2vec, test_loader_word2vec, loss_fn)

    # 기록 저장
    train_losses.append(train_loss)
    val_losses.append(val_loss)
    current_lr = optimizer.param_groups[0]['lr']
    learning_rates.append(current_lr)

    # 스케줄러 업데이트
    scheduler.step(val_loss)

    print(f"Epoch {epoch+1}, Train Loss: {train_loss:.4f}, Val Loss: {val_loss:.4f}, LR: {current_lr:.6f}")

# 그래프 시각화
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# 1. 손실함수 그래프
axes[0].plot(train_losses, label='Train Loss', marker='o', markersize=3)
axes[0].plot(val_losses, label='Validation Loss', marker='s', markersize=3)
axes[0].set_xlabel('Epoch')
axes[0].set_ylabel('Loss')
axes[0].set_title('Training and Validation Loss')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# 2. 학습률 그래프
axes[1].plot(learning_rates, label='Learning Rate', marker='o', markersize=3, color='green')
axes[1].set_xlabel('Epoch')
axes[1].set_ylabel('Learning Rate')
axes[1].set_title('Learning Rate Schedule')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

predictions, labels = evaluate(model_word2vec, test_loader_word2vec)
````
